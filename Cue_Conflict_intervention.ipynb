```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# Paths
original_path = "/content/AI_Mistakes_Error_Framework/stimuli_original_novel/geirhos-cue-conflict/airplane/airplane4-keyboard3.png"
mask_path = "/content/developmental-shape-bias/stimuli/geirhos-masks/airplane/airplane4.png"

# Load
original = np.array(Image.open(original_path).convert("RGB")).astype(np.float32)
mask = np.array(Image.open(mask_path).convert("L")).astype(np.float32)

# Object weight:
# black mask = object
# white mask = background
object_weight = 1.0 - (mask / 255.0)

# Background weight
background_weight = mask / 255.0

# Alpha levels
alpha_levels = [0.00, 0.25, 0.50, 0.75, 1.00]

generated = {}

for alpha in alpha_levels:

    # Amount of background that remains
    background_remaining = 1.0 - alpha

    # Keep object unchanged.
    # Gradually replace background with white.
    output = (
        original * object_weight[:, :, None]
        +
        (
            original * background_weight[:, :, None] * background_remaining
            +
            255.0 * background_weight[:, :, None] * alpha
        )
    )

    output = np.clip(output, 0, 255).astype(np.uint8)

    generated[alpha] = Image.fromarray(output)


# Display
fig, axes = plt.subplots(1, 5, figsize=(20, 4))

for ax, alpha in zip(axes, alpha_levels):
    ax.imshow(generated[alpha])
    ax.set_title(f"α = {alpha}")
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# CUE-CONFLICT ALPHA INTERVENTION
# COMPLETE SETUP + TEST + VALIDATION
# ============================================================

import os
import subprocess
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image


# ============================================================
# 1. CLONE REPOSITORIES IF NEEDED
# ============================================================

darshini_repo = Path("/content/AI_Mistakes_Error_Framework")
tartaglini_repo = Path("/content/developmental-shape-bias")

if not darshini_repo.exists():
    print("Cloning Darshini repository...")
    subprocess.run(
        [
            "git", "clone",
            "https://github.com/Darshini-CogSci/AI_Mistakes_Error_Framework.git",
            str(darshini_repo)
        ],
        check=True
    )
else:
    print("Darshini repository already exists.")

if not tartaglini_repo.exists():
    print("Cloning Tartaglini repository...")
    subprocess.run(
        [
            "git", "clone",
            "https://github.com/alexatartaglini/developmental-shape-bias.git",
            str(tartaglini_repo)
        ],
        check=True
    )
else:
    print("Tartaglini repository already exists.")


# ============================================================
# 2. DEFINE IMPORTANT DIRECTORIES
# ============================================================

stimulus_root = (
    darshini_repo /
    "stimuli_original_novel"
)

cue_conflict_root = (
    stimulus_root /
    "geirhos-cue-conflict"
)

clean_cue_root = (
    stimulus_root /
    "tartaglini_clean_cue_conflict"
)

mask_root = (
    tartaglini_repo /
    "stimuli" /
    "geirhos-masks"
)

tartaglini_stimuli_root = (
    tartaglini_repo /
    "stimuli"
)


# ============================================================
# 3. CHECK REPOSITORY STRUCTURE
# ============================================================

print("\n" + "=" * 60)
print("REPOSITORY CHECK")
print("=" * 60)

print("\nDarshini stimulus folder:")
print(os.listdir(stimulus_root))

print("\nTartaglini stimulus folders containing 'geirhos':")

for folder in os.listdir(tartaglini_stimuli_root):
    if "geirhos" in folder.lower():
        print("  ", folder)


# ============================================================
# 4. SELECT ONE TEST IMAGE
# ============================================================

category = "airplane"
filename = "airplane4-keyboard3.png"

original_path = (
    cue_conflict_root /
    category /
    filename
)

clean_path = (
    clean_cue_root /
    category /
    filename
)

# Corresponding Geirhos object mask
mask_path = (
    mask_root /
    category /
    "airplane4.png"
)


print("\n" + "=" * 60)
print("TEST IMAGE")
print("=" * 60)

print("Original:")
print(original_path)

print("\nClean:")
print(clean_path)

print("\nMask:")
print(mask_path)


# ============================================================
# 5. VERIFY FILES EXIST
# ============================================================

for path in [original_path, clean_path, mask_path]:
    if not path.exists():
        raise FileNotFoundError(
            f"\nRequired file was not found:\n{path}"
        )

print("\nAll required files found.")


# ============================================================
# 6. LOAD IMAGES
# ============================================================

original_img = Image.open(original_path).convert("RGB")
clean_img = Image.open(clean_path).convert("RGB")
mask_img = Image.open(mask_path).convert("L")


# ============================================================
# 7. BASIC IMAGE / MASK VALIDATION
# ============================================================

print("\n" + "=" * 60)
print("IMAGE / MASK VALIDATION")
print("=" * 60)

print("Original size:", original_img.size)
print("Clean size:", clean_img.size)
print("Mask size:", mask_img.size)
print("Mask mode:", mask_img.mode)

if original_img.size != (224, 224):
    raise ValueError("Original image is not 224 × 224.")

if mask_img.size != original_img.size:
    raise ValueError(
        "Mask and original image do not have the same dimensions."
    )

print("\nImage and mask dimensions match.")


# ============================================================
# 8. MASK STATISTICS
# ============================================================

original_array = np.array(original_img).astype(np.float32)
clean_array = np.array(clean_img).astype(np.float32)
mask_array = np.array(mask_img).astype(np.float32)

unique_values = np.unique(mask_array)

zero_pixels = np.sum(mask_array == 0)
full_pixels = np.sum(mask_array == 255)
intermediate_pixels = np.sum(
    (mask_array > 0) &
    (mask_array < 255)
)

total_pixels = mask_array.size

print("\nMask statistics:")
print("Total pixels:", total_pixels)
print("Pixels exactly 0:", zero_pixels)
print("Pixels exactly 255:", full_pixels)
print("Intermediate pixels:", intermediate_pixels)
print("Number of unique mask values:", len(unique_values))


# ============================================================
# 9. DISPLAY ORIGINAL, CLEAN IMAGE AND MASK
# ============================================================

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(original_img)
axes[0].set_title("Original cue-conflict")
axes[0].axis("off")

axes[1].imshow(clean_img)
axes[1].set_title("Clean cue-conflict")
axes[1].axis("off")

axes[2].imshow(mask_img, cmap="gray", vmin=0, vmax=255)
axes[2].set_title("Geirhos object mask")
axes[2].axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# 10. CREATE OUTPUT DIRECTORY
# ============================================================

output_root = Path("/content/cue_conflict_alpha_test")
output_root.mkdir(parents=True, exist_ok=True)

print("\nOutput directory:")
print(output_root)


# ============================================================
# 11. PREPARE MASK WEIGHTS
# ============================================================

# Black mask = object
# White mask = background
#
# Convert the grayscale mask into continuous weights.
#
# Object weight:
#   mask = 0   -> object weight = 1
#   mask = 255 -> object weight = 0
#
# Background weight:
#   mask = 0   -> background weight = 0
#   mask = 255 -> background weight = 1
#
# Intermediate values preserve the anti-aliased boundary.

object_weight = 1.0 - (mask_array / 255.0)
background_weight = mask_array / 255.0


# ============================================================
# 12. DEFINE ALPHA LEVELS
# ============================================================

alpha_levels = [
    0.00,
    0.25,
    0.50,
    0.75,
    1.00
]


# ============================================================
# 13. GENERATE FIVE TEST STIMULI
# ============================================================

generated = {}

for alpha in alpha_levels:

    # Background remaining:
    #
    # alpha = 0.00 -> 100% background
    # alpha = 0.25 -> 75% background
    # alpha = 0.50 -> 50% background
    # alpha = 0.75 -> 25% background
    # alpha = 1.00 -> 0% background

    background_remaining = 1.0 - alpha

    # Keep the object fixed.
    object_part = (
        original_array *
        object_weight[:, :, None]
    )

    # Retain only the desired amount of background.
    background_part = (
        original_array *
        background_weight[:, :, None] *
        background_remaining
    )

    # Replace the removed background with white.
    white_part = (
        255.0 *
        background_weight[:, :, None] *
        alpha
    )

    # Combine all components.
    output = (
        object_part +
        background_part +
        white_part
    )

    output = np.clip(output, 0, 255).astype(np.uint8)

    generated[alpha] = Image.fromarray(output)


# ============================================================
# 14. SAVE THE FIVE TEST STIMULI
# ============================================================

for alpha, image in generated.items():

    alpha_name = str(alpha).replace(".", "_")

    save_path = (
        output_root /
        f"{filename[:-4]}_alpha_{alpha_name}.png"
    )

    image.save(save_path)

    print(
        f"Saved α={alpha:.2f}: {save_path}"
    )


# ============================================================
# 15. DISPLAY THE FIVE ALPHA CONDITIONS
# ============================================================

fig, axes = plt.subplots(
    1,
    len(alpha_levels),
    figsize=(20, 4)
)

for ax, alpha in zip(axes, alpha_levels):

    ax.imshow(generated[alpha])

    ax.set_title(
        f"α = {alpha}\n"
        f"Background = {(1-alpha)*100:.0f}%"
    )

    ax.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# 16. IMPORTANT VALIDATION:
#     DOES α=0 REPRODUCE THE ORIGINAL?
# ============================================================

original_uint8 = np.array(
    Image.open(original_path).convert("RGB")
)

alpha0_uint8 = np.array(
    generated[0.00]
)

difference = np.abs(
    original_uint8.astype(int) -
    alpha0_uint8.astype(int)
)

maximum_difference = difference.max()
mean_difference = difference.mean()

changed_pixels = np.sum(
    np.any(difference != 0, axis=2)
)

print("\n" + "=" * 60)
print("ALPHA = 0 VALIDATION")
print("=" * 60)

print("Maximum pixel difference:", maximum_difference)
print("Mean pixel difference:", mean_difference)
print("Pixels changed:", changed_pixels)

if maximum_difference == 0:
    print("\nSUCCESS: α=0 reproduces the original image exactly.")
else:
    print(
        "\nNOTE: α=0 does not reproduce the original exactly."
    )


# ============================================================
# 17. CHECK THAT IMAGE SIZE REMAINS CONSTANT
# ============================================================

print("\n" + "=" * 60)
print("OUTPUT SIZE CHECK")
print("=" * 60)

for alpha, image in generated.items():

    print(
        f"α={alpha:.2f}: "
        f"{image.size} | "
        f"{image.mode}"
    )


# ============================================================
# 18. SAVE VALIDATION REPORT
# ============================================================

report_path = output_root / "validation_report.txt"

with open(report_path, "w") as f:

    f.write("CUE-CONFLICT ALPHA INTERVENTION TEST\n")
    f.write("=" * 50 + "\n\n")

    f.write(f"Category: {category}\n")
    f.write(f"Filename: {filename}\n\n")

    f.write("Original image:\n")
    f.write(f"{original_path}\n\n")

    f.write("Mask:\n")
    f.write(f"{mask_path}\n\n")

    f.write("Image size:\n")
    f.write(f"{original_img.size}\n\n")

    f.write("Mask statistics:\n")
    f.write(f"Total pixels: {total_pixels}\n")
    f.write(f"Pixels exactly 0: {zero_pixels}\n")
    f.write(f"Pixels exactly 255: {full_pixels}\n")
    f.write(f"Intermediate pixels: {intermediate_pixels}\n")
    f.write(f"Unique mask values: {len(unique_values)}\n\n")

    f.write("Alpha levels:\n")

    for alpha in alpha_levels:
        f.write(
            f"alpha={alpha:.2f}, "
            f"background remaining={(1-alpha)*100:.0f}%\n"
        )

    f.write("\nAlpha=0 validation:\n")
    f.write(
        f"Maximum pixel difference: "
        f"{maximum_difference}\n"
    )

    f.write(
        f"Mean pixel difference: "
        f"{mean_difference}\n"
    )

    f.write(
        f"Pixels changed: "
        f"{changed_pixels}\n"
    )


print("\nValidation report saved:")
print(report_path)


# ============================================================
# 19. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("DONE")
print("=" * 60)

print("""
We have now:

1. Located the Darshini cue-conflict stimuli.
2. Located the Tartaglini repository.
3. Verified the Geirhos mask.
4. Confirmed image and mask are both 224 × 224.
5. Confirmed the mask represents the object against the background.
6. Preserved the grayscale/anti-aliased mask.
7. Created five alpha conditions:
      α=0.00
      α=0.25
      α=0.50
      α=0.75
      α=1.00
8. Saved the five test images.
9. Tested whether α=0 reproduces the original exactly.
10. Saved a validation report.

IMPORTANT:
This has only generated the TEST stimulus for
airplane4-keyboard3.png.

We have NOT yet generated the 51 × 5 = 255
experimental stimuli.
""")
```

```python
# ============================================================
# CUE-CONFLICT INTERVENTION
# STAGE 1: GENERATE + VALIDATE ALL 255 STIMULI
# ============================================================

import os
import glob
import numpy as np
import pandas as pd
from PIL import Image

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

DARSHINI_ROOT = "/content/AI_Mistakes_Error_Framework"
TARTAGLINI_ROOT = "/content/developmental-shape-bias"

STIMULI_ROOT = os.path.join(
    DARSHINI_ROOT,
    "stimuli_original_novel"
)

CUE_CONFLICT_ROOT = os.path.join(
    STIMULI_ROOT,
    "geirhos-cue-conflict"
)

MASTER_TABLE = os.path.join(
    DARSHINI_ROOT,
    "false_knife_master_table.csv"
)

MASK_ROOT = os.path.join(
    TARTAGLINI_ROOT,
    "stimuli",
    "geirhos-masks"
)

OUTPUT_ROOT = "/content/cue_conflict_alpha_intervention"

os.makedirs(OUTPUT_ROOT, exist_ok=True)

print("=" * 70)
print("CUE-CONFLICT INTERVENTION")
print("STAGE 1: GENERATING EXPERIMENTAL STIMULI")
print("=" * 70)


# ------------------------------------------------------------
# 2. CHECK REQUIRED PATHS
# ------------------------------------------------------------

required_paths = {
    "Cue-conflict stimulus folder": CUE_CONFLICT_ROOT,
    "Master table": MASTER_TABLE,
    "Geirhos mask folder": MASK_ROOT
}

print("\nPATH CHECK")
print("-" * 70)

for name, path in required_paths.items():
    exists = os.path.exists(path)
    print(f"{name}: {'FOUND' if exists else 'MISSING'}")
    print(path)

    if not exists:
        raise FileNotFoundError(f"\nRequired path not found:\n{path}")


# ------------------------------------------------------------
# 3. LOAD MASTER TABLE
# ------------------------------------------------------------

df = pd.read_csv(MASTER_TABLE)

print("\nMASTER TABLE")
print("-" * 70)

print("Rows:", len(df))
print("Columns:")
print(df.columns.tolist())


# ------------------------------------------------------------
# 4. INSPECT CONDITION VALUES
# ------------------------------------------------------------

print("\nCONDITION VALUES")
print("-" * 70)

if "condition" in df.columns:
    print(df["condition"].value_counts(dropna=False))
else:
    print("WARNING: No 'condition' column found.")


# ------------------------------------------------------------
# 5. IDENTIFY CUE-CONFLICT FALSE-KNIFE CASES
# ------------------------------------------------------------

# We expect the master table to contain the cue-conflict
# condition and the false-knife cases.

cue_df = df.copy()

if "condition" in cue_df.columns:
    cue_df = cue_df[
        cue_df["condition"].astype(str).str.lower().str.contains(
            "cue", na=False
        )
    ].copy()

print("\nCUE-CONFLICT CASES")
print("-" * 70)
print("Number of cue-conflict rows:", len(cue_df))

# ------------------------------------------------------------
# 6. FILTER FALSE-KNIFE CASES
# ------------------------------------------------------------

# The project definition:
# actual category != knife AND model response == knife

if "model_response" in cue_df.columns and "actual_on_object" in cue_df.columns:

    cue_df = cue_df[
        (cue_df["model_response"].astype(str).str.lower() == "knife") &
        (cue_df["actual_on_object"].astype(str).str.lower() != "knife")
    ].copy()

elif "model_response" in cue_df.columns and "object_response" in cue_df.columns:

    cue_df = cue_df[
        (cue_df["model_response"].astype(str).str.lower() == "knife") &
        (cue_df["object_response"].astype(str).str.lower() != "knife")
    ].copy()

else:
    raise ValueError(
        "Could not identify the required category columns in the master table."
    )

print("False-knife cue-conflict cases:", len(cue_df))

if len(cue_df) != 51:
    print("\nWARNING:")
    print("Expected 51 cue-conflict false-knife cases.")
    print("Found:", len(cue_df))
    print("We will STOP here rather than generating the wrong 255 stimuli.")

    print("\nAvailable columns:")
    print(cue_df.columns.tolist())

    raise ValueError(
        f"Expected 51 cue-conflict false-knife cases, found {len(cue_df)}."
    )


# ------------------------------------------------------------
# 7. FIND THE IMAGE COLUMN
# ------------------------------------------------------------

possible_image_columns = [
    "image",
    "image_name",
    "filename",
    "file_name",
    "stimulus",
    "stimulus_name",
    "original_path"
]

image_column = None

for col in possible_image_columns:
    if col in cue_df.columns:
        image_column = col
        break

if image_column is None:
    raise ValueError(
        "Could not identify the image filename/path column."
    )

print("\nIMAGE COLUMN")
print("-" * 70)
print("Using:", image_column)


# ------------------------------------------------------------
# 8. HELPER TO FIND ORIGINAL IMAGE
# ------------------------------------------------------------

def find_original_image(value):

    value = str(value)

    # If the table already contains a full path
    if os.path.isfile(value):
        return value

    # If it contains a path relative to the stimulus folder
    candidate = os.path.join(CUE_CONFLICT_ROOT, value)

    if os.path.isfile(candidate):
        return candidate

    # Search by filename
    filename = os.path.basename(value)

    matches = glob.glob(
        os.path.join(
            CUE_CONFLICT_ROOT,
            "*",
            filename
        )
    )

    if len(matches) == 1:
        return matches[0]

    if len(matches) > 1:
        print("Multiple matches found for:", filename)
        print(matches)
        return matches[0]

    return None


# ------------------------------------------------------------
# 9. HELPER TO FIND CORRESPONDING MASK
# ------------------------------------------------------------

def find_mask(original_path):

    filename = os.path.basename(original_path)

    # Example:
    # airplane4-keyboard3.png
    #
    # Corresponding object mask:
    # airplane4.png

    stem = os.path.splitext(filename)[0]

    # Extract object ID before the first "-"
    object_id = stem.split("-")[0]

    # Example:
    # airplane4

    category = os.path.basename(
        os.path.dirname(original_path)
    )

    mask_path = os.path.join(
        MASK_ROOT,
        category,
        object_id + ".png"
    )

    if os.path.isfile(mask_path):
        return mask_path

    return None


# ------------------------------------------------------------
# 10. ALPHA LEVELS
# ------------------------------------------------------------

ALPHA_LEVELS = [0.00, 0.25, 0.50, 0.75, 1.00]

print("\nALPHA LEVELS")
print("-" * 70)
print(ALPHA_LEVELS)


# ------------------------------------------------------------
# 11. PREPARE OUTPUT DIRECTORIES
# ------------------------------------------------------------

for alpha in ALPHA_LEVELS:

    alpha_str = str(alpha).replace(".", "_")

    os.makedirs(
        os.path.join(
            OUTPUT_ROOT,
            f"alpha_{alpha_str}"
        ),
        exist_ok=True
    )


# ------------------------------------------------------------
# 12. GENERATE STIMULI
# ------------------------------------------------------------

records = []

missing_images = []
missing_masks = []

total_generated = 0

print("\nGENERATING STIMULI")
print("-" * 70)

for index, row in cue_df.reset_index(drop=True).iterrows():

    image_value = row[image_column]

    original_path = find_original_image(image_value)

    if original_path is None:

        missing_images.append(str(image_value))
        continue

    mask_path = find_mask(original_path)

    if mask_path is None:

        missing_masks.append(original_path)
        continue

    # --------------------------------------------------------
    # LOAD ORIGINAL + MASK
    # --------------------------------------------------------

    original = Image.open(original_path).convert("RGB")
    mask = Image.open(mask_path).convert("L")

    original_array = np.array(original).astype(np.float32)
    mask_array = np.array(mask).astype(np.float32)

    # Safety check
    if original_array.shape[:2] != mask_array.shape:

        raise ValueError(
            f"Image/mask size mismatch:\n"
            f"{original_path}\n"
            f"Image: {original_array.shape}\n"
            f"Mask: {mask_array.shape}"
        )

    # --------------------------------------------------------
    # NORMALIZED MASK
    # --------------------------------------------------------

    # Mask convention:
    # 0   = object
    # 255 = background
    #
    # Therefore:
    # object_weight = 1 at object
    # background_weight = 1 at background

    object_weight = 1.0 - (mask_array / 255.0)
    background_weight = mask_array / 255.0

    filename = os.path.basename(original_path)

    stem = os.path.splitext(filename)[0]

    category = os.path.basename(
        os.path.dirname(original_path)
    )

    # --------------------------------------------------------
    # GENERATE FIVE ALPHA CONDITIONS
    # --------------------------------------------------------

    for alpha in ALPHA_LEVELS:

        alpha_str = str(alpha).replace(".", "_")

        output_dir = os.path.join(
            OUTPUT_ROOT,
            f"alpha_{alpha_str}"
        )

        output_filename = (
            f"{stem}_alpha_{alpha_str}.png"
        )

        output_path = os.path.join(
            output_dir,
            output_filename
        )

        # ----------------------------------------------------
        # α = 0
        # EXACT COPY OF ORIGINAL
        # ----------------------------------------------------

        if alpha == 0.00:

            output = original.copy()

        else:

            # ------------------------------------------------
            # Keep object fixed.
            #
            # Fade only the background toward white.
            # ------------------------------------------------

            object_part = (
                original_array *
                object_weight[:, :, None]
            )

            background_part = (
                original_array *
                background_weight[:, :, None] *
                (1.0 - alpha)
            )

            white_part = (
                255.0 *
                background_weight[:, :, None] *
                alpha
            )

            output_array = (
                object_part +
                background_part +
                white_part
            )

            output_array = np.clip(
                output_array,
                0,
                255
            ).round().astype(np.uint8)

            output = Image.fromarray(
                output_array,
                mode="RGB"
            )

        output.save(output_path)

        # ----------------------------------------------------
        # RECORD METADATA
        # ----------------------------------------------------

        records.append({
            "image_id": stem,
            "filename": filename,
            "category": category,
            "original_path": original_path,
            "mask_path": mask_path,
            "alpha": alpha,
            "output_path": output_path,
            "is_baseline": alpha == 0.00
        })

        total_generated += 1


# ------------------------------------------------------------
# 13. REPORT MISSING FILES
# ------------------------------------------------------------

print("\nFILE CHECK")
print("-" * 70)

print("Missing original images:", len(missing_images))
print("Missing masks:", len(missing_masks))

if missing_images:

    print("\nMissing images:")
    for x in missing_images:
        print(x)

if missing_masks:

    print("\nMissing masks:")
    for x in missing_masks:
        print(x)

if missing_images or missing_masks:

    raise ValueError(
        "Some required files were missing. "
        "No final experimental analysis should be performed yet."
    )


# ------------------------------------------------------------
# 14. SAVE MASTER STIMULUS CSV
# ------------------------------------------------------------

stimulus_df = pd.DataFrame(records)

csv_path = os.path.join(
    OUTPUT_ROOT,
    "cue_conflict_alpha_stimulus_master.csv"
)

stimulus_df.to_csv(
    csv_path,
    index=False
)


# ------------------------------------------------------------
# 15. BASIC COUNT VALIDATION
# ------------------------------------------------------------

print("\nSTIMULUS COUNT VALIDATION")
print("-" * 70)

print("False-knife images:", cue_df.shape[0])
print("Alpha levels:", len(ALPHA_LEVELS))
print("Expected stimuli:", 51 * 5)
print("Generated stimuli:", total_generated)

if total_generated != 255:

    raise ValueError(
        f"Expected 255 stimuli but generated {total_generated}."
    )

print("COUNT CHECK: PASSED")


# ------------------------------------------------------------
# 16. CHECK EACH ALPHA LEVEL
# ------------------------------------------------------------

print("\nALPHA COUNT")
print("-" * 70)

print(
    stimulus_df["alpha"]
    .value_counts()
    .sort_index()
)


# ------------------------------------------------------------
# 17. CHECK EACH IMAGE HAS FIVE CONDITIONS
# ------------------------------------------------------------

counts_per_image = (
    stimulus_df
    .groupby("image_id")["alpha"]
    .nunique()
)

print("\nCONDITIONS PER IMAGE")
print("-" * 70)

print(counts_per_image.value_counts())

if not (counts_per_image == 5).all():

    raise ValueError(
        "At least one image does not have all five alpha conditions."
    )

print("PER-IMAGE CHECK: PASSED")


# ------------------------------------------------------------
# 18. EXACT α=0 VALIDATION
# ------------------------------------------------------------

print("\nα = 0 BASELINE VALIDATION")
print("-" * 70)

baseline_df = stimulus_df[
    stimulus_df["alpha"] == 0.00
]

max_difference_overall = 0
total_changed_pixels = 0

for _, row in baseline_df.iterrows():

    original = np.array(
        Image.open(row["original_path"]).convert("RGB")
    ).astype(np.int16)

    generated = np.array(
        Image.open(row["output_path"]).convert("RGB")
    ).astype(np.int16)

    difference = np.abs(
        original - generated
    )

    max_difference = difference.max()

    changed_pixels = np.sum(
        np.any(difference != 0, axis=2)
    )

    max_difference_overall = max(
        max_difference_overall,
        max_difference
    )

    total_changed_pixels += changed_pixels


print("Maximum pixel difference:", max_difference_overall)
print("Total changed pixels across all α=0 images:",
      total_changed_pixels)

if max_difference_overall != 0:

    raise ValueError(
        "α=0 baseline is not exactly identical to the original."
    )

print("EXACT BASELINE CHECK: PASSED")


# ------------------------------------------------------------
# 19. IMAGE SIZE / MODE CHECK
# ------------------------------------------------------------

print("\nIMAGE FORMAT VALIDATION")
print("-" * 70)

format_errors = []

for _, row in stimulus_df.iterrows():

    img = Image.open(row["output_path"])

    if img.size != (224, 224):
        format_errors.append(
            (row["output_path"], "wrong size", img.size)
        )

    if img.mode != "RGB":
        format_errors.append(
            (row["output_path"], "wrong mode", img.mode)
        )


print("Format errors:", len(format_errors))

if format_errors:

    for error in format_errors[:10]:
        print(error)

    raise ValueError(
        "Some generated images have incorrect size or mode."
    )

print("FORMAT CHECK: PASSED")


# ------------------------------------------------------------
# 20. CHECK α=1 BACKGROUND SUPPRESSION
# ------------------------------------------------------------

print("\nα = 1 BACKGROUND CHECK")
print("-" * 70)

# We are not requiring the α=1 image to equal the existing
# 'clean cue-conflict' image.
#
# We only check that α=1 differs from α=0 for the expected
# background-containing images.

alpha0_df = stimulus_df[
    stimulus_df["alpha"] == 0.00
].set_index("image_id")

alpha1_df = stimulus_df[
    stimulus_df["alpha"] == 1.00
].set_index("image_id")

different_images = 0

for image_id in alpha0_df.index:

    img0 = np.array(
        Image.open(alpha0_df.loc[image_id, "output_path"])
    ).astype(np.int16)

    img1 = np.array(
        Image.open(alpha1_df.loc[image_id, "output_path"])
    ).astype(np.int16)

    if not np.array_equal(img0, img1):
        different_images += 1


print(
    "Images where α=1 differs from α=0:",
    different_images,
    "/",
    len(alpha0_df)
)


# ------------------------------------------------------------
# 21. SAVE VALIDATION REPORT
# ------------------------------------------------------------

report_path = os.path.join(
    OUTPUT_ROOT,
    "cue_conflict_alpha_validation_report.txt"
)

with open(report_path, "w") as f:

    f.write(
        "CUE-CONFLICT ALPHA INTERVENTION\n"
        "STAGE 1 VALIDATION REPORT\n"
        "========================================\n\n"
    )

    f.write(
        f"False-knife images: {len(cue_df)}\n"
    )

    f.write(
        f"Alpha levels: {ALPHA_LEVELS}\n"
    )

    f.write(
        f"Expected stimuli: 255\n"
    )

    f.write(
        f"Generated stimuli: {total_generated}\n\n"
    )

    f.write(
        "Alpha=0 exact baseline check: PASSED\n"
    )

    f.write(
        f"Maximum alpha=0 pixel difference: "
        f"{max_difference_overall}\n"
    )

    f.write(
        f"Total changed alpha=0 pixels: "
        f"{total_changed_pixels}\n\n"
    )

    f.write(
        "Image format check: PASSED\n"
    )

    f.write(
        "All images: 224x224 RGB\n"
    )

    f.write(
        f"Images changed at alpha=1: "
        f"{different_images}/51\n"
    )


# ------------------------------------------------------------
# 22. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STAGE 1 COMPLETE")
print("=" * 70)

print("\nGenerated:", total_generated, "stimuli")
print("Expected:", 255)

print("\nOutput folder:")
print(OUTPUT_ROOT)

print("\nMaster CSV:")
print(csv_path)

print("\nValidation report:")
print(report_path)

print("\nThe 255 stimuli are ready for inspection.")
print("DO NOT RUN THE RESNET PREDICTION YET.")

print("=" * 70)
```

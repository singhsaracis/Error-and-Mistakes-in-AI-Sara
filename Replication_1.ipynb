```python
import pandas as pd
import os

trial_path = "/content/drive/MyDrive/AI_Mistakes_Error_Framework/trial_data"

all_data = []

for file in os.listdir(trial_path):

    if file.endswith(".csv"):

        temp = pd.read_csv(
            os.path.join(trial_path, file)
        )

        temp["source_file"] = file

        all_data.append(temp)

df_all = pd.concat(
    all_data,
    ignore_index=True
)

print("Shape:", df_all.shape)

print("\nConditions/files:")

print(
    df_all["source_file"].unique()
)
```

```python
knife_bias = []

for file in df_all["source_file"].unique():

    subset = df_all[
        df_all["source_file"] == file
    ]

    non_knife_images = subset[
        subset["category"] != "knife"
    ]

    false_knives = non_knife_images[
        non_knife_images["object_response"] == "knife"
    ]

    knife_percent = (
        len(false_knives)
        /
        len(non_knife_images)
    ) * 100

    knife_bias.append(
        [file, round(knife_percent,2)]
    )

knife_df = pd.DataFrame(
    knife_bias,
    columns=[
        "Condition",
        "Knife_Bias_Percent"
    ]
)

knife_df = knife_df.sort_values(
    "Knife_Bias_Percent",
    ascending=False
)

print(knife_df)
```

```python
# %load /content/drive/MyDrive/AI_Mistakes_Error_Framework/extract_layer3_embeddings.py
"""
Extracts penultimate layer embeddings from ResNet50 for a folder of images.

Folder structure expected:
  images/
    airplane/
      img1.png
      img2.png
    bear/
      ...
    ...  (any of the 16 Geirhos categories)

Output:
  normal_images_embeddings.json  <- {filename: [2048-dim vector]}
  normal_images_shape_classes.json <- {filename: {shape, category, dir}}
"""

import os
import json
import torch
import torch.nn as nn
from torchvision import models, transforms
from PIL import Image
from collections import defaultdict

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Using device: {device}')

# ── Configuration ──────────────────────────────────────────────────────────────
IMAGE_DIR = "/content/drive/MyDrive/AI_Mistakes_Error_Framework/stimuli_original_novel/novel_noise_filled"

OUTPUT_EMB = "/content/drive/MyDrive/AI_Mistakes_Error_Framework/layer3_noise_embeddings.json"

OUTPUT_SC = "/content/drive/MyDrive/AI_Mistakes_Error_Framework/layer3_noise_shape_classes.json"
BATCH_SIZE  = 32

CATEGORIES = [
    'airplane', 'bear', 'bicycle', 'bird', 'boat', 'bottle',
    'car', 'cat', 'chair', 'clock', 'dog', 'elephant',
    'keyboard', 'knife', 'oven', 'truck'
]

# ── Load model ─────────────────────────────────────────────────────────────────
print('Loading ResNet50...')
model = models.resnet50(pretrained=True)
model.eval()

# Strip final FC layer — same as main.py
modules = list(model.children())[:7]

layer3_model = nn.Sequential(*modules).to(device)

layer3_model.eval()

# ImageNet transforms — same as main.py
transform = transforms.Compose([
    transforms.Resize(224),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225]),
])

# ── Collect image paths ────────────────────────────────────────────────────────
image_paths = []
shape_classes = {}

for cat in CATEGORIES:
    cat_dir = os.path.join(IMAGE_DIR, cat)
    if not os.path.exists(cat_dir):
        print(f'  WARNING: {cat_dir} not found, skipping')
        continue
    for fname in sorted(os.listdir(cat_dir)):
        if not fname.lower().endswith(('.png', '.jpg', '.jpeg')):
            continue
        full_path = os.path.join(cat_dir, fname)
        rel_dir   = os.path.join(cat, fname)
        image_paths.append((fname, full_path, cat, rel_dir))
        shape_classes[fname] = {
            'shape':    cat,
            'category': cat,
            'dir':      rel_dir,
            'type':     'normal'
        }

print(f'Found {len(image_paths)} images across {len(set(p[2] for p in image_paths))} categories')

# ── Extract embeddings in batches ─────────────────────────────────────────────
embedding_dict = {}

with torch.no_grad():
    for i in range(0, len(image_paths), BATCH_SIZE):
        batch_paths = image_paths[i:i+BATCH_SIZE]
        batch_imgs  = []
        batch_names = []

        for fname, full_path, cat, rel_dir in batch_paths:
            try:
                img = Image.open(full_path).convert('RGB')
                img = transform(img)
                batch_imgs.append(img)
                batch_names.append(fname)
            except Exception as e:
                print(f'  ERROR loading {full_path}: {e}')

        if not batch_imgs:
            continue

        batch_tensor = torch.stack(batch_imgs).to(device)
        embeddings = layer3_model(batch_tensor)
        embeddings = torch.mean(embeddings, dim=[2,3])
        embeddings   = torch.squeeze(embeddings)

        if embeddings.dim() == 1:
            embeddings = embeddings.unsqueeze(0)

        for j, name in enumerate(batch_names):
            embedding_dict[name] = embeddings[j].cpu().tolist()

        print(f'  Processed {min(i+BATCH_SIZE, len(image_paths))}/{len(image_paths)}')

# ── Save ───────────────────────────────────────────────────────────────────────
os.makedirs(os.path.dirname(OUTPUT_EMB), exist_ok=True)
os.makedirs(os.path.dirname(OUTPUT_SC),  exist_ok=True)

with open(OUTPUT_EMB, 'w') as f:
    json.dump(embedding_dict, f)
print(f'\nSaved {len(embedding_dict)} embeddings -> {OUTPUT_EMB}')

with open(OUTPUT_SC, 'w') as f:
    json.dump(shape_classes, f, indent=2)
print(f'Saved shape_classes -> {OUTPUT_SC}')
```

```python
import os

base = "/content/drive/MyDrive/AI_Mistakes_Error_Framework/tsne of different layers"

files_to_fix = {

    "conv1_knife_tsne.py": {

        "old": "CONV1_FILES = {",
        "new_dict": """CONV1_FILES = {

    "silhouette":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/conv1_silhouette_embeddings.json",

    "edges":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/conv1_edges_embeddings.json",

    "outline":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/conv1_outline_embeddings.json",

    "noise_filled":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/conv1_noise_embeddings.json"
}"""
    },

    "layer2_knife_tsne.py": {

        "old": "LAYER2_FILES = {",
        "new_dict": """LAYER2_FILES = {

    "silhouette":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/layer2_silhouette_embeddings.json",

    "edges":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/layer2_edges_embeddings.json",

    "outline":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/layer2_outline_embeddings.json",

    "noise_filled":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/layer2_noise_embeddings.json"
}"""
    },

    "avgpool_knife_tsne.py": {

        "old": "AVGPOOL_FILES = {",
        "new_dict": """AVGPOOL_FILES = {

    "silhouette":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/avgpool_silhouette_embeddings.json",

    "edges":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/avgpool_edges_embeddings.json",

    "outline":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/avgpool_outline_embeddings.json",

    "noise_filled":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/avgpool_noise_embeddings.json"
}"""
    }
}

for fname, cfg in files_to_fix.items():

    path = os.path.join(base, fname)

    with open(path, "r") as f:
        lines = f.readlines()

    new_lines = []

    inside_dict = False

    for line in lines:

        stripped = line.strip()

        if stripped.startswith(cfg["old"]):

            inside_dict = True
            new_lines.append(cfg["new_dict"] + "\n")
            continue

        if inside_dict:

            if stripped == "}":
                inside_dict = False

            continue

        new_lines.append(line)

    with open(path, "w") as f:
        f.writelines(new_lines)

    print("Fixed:", fname)

print("\nDone patching all scripts.")
```

```python
from google.colab import drive
drive.mount('/content/drive')

import json
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

from sklearn.manifold import TSNE
from sklearn.decomposition import PCA

# ----------------------------
# PATHS
# ----------------------------

BASE_DIR = "/content/drive/MyDrive/AI_Mistakes_Error_Framework"

CONV1_FILES = {
    "silhouette": f"{BASE_DIR}/conv1_silhouette_embeddings.json",
    "edges": f"{BASE_DIR}/conv1_edges_embeddings.json",
    "outline": f"{BASE_DIR}/conv1_outline_embeddings.json",
    "noise_filled": f"{BASE_DIR}/conv1_noise_embeddings.json"
}

TRIAL_CSVS = {
    "silhouette": f"{BASE_DIR}/trial_data/black_filled_silhouettes_trial_data.csv",
    "edges": f"{BASE_DIR}/trial_data/original_edges_trial_data_FIXED.csv",
    "outline": f"{BASE_DIR}/trial_data/outlines_edges_trial_data.csv",
    "noise_filled": f"{BASE_DIR}/trial_data/noise_filled_silhouettes_trial_data.csv"
}

# ----------------------------
# HELPERS
# ----------------------------

def clean_name(name):
    name = str(name).lower().split("/")[-1]
    for r in [".png",".jpg",".jpeg","_edges","_outline","_hq_outline","_noise_filled","_hq","edges_","outline_","noise_"]:
        name = name.replace(r,"")
    return name

def get_true_category(name):
    stem = clean_name(name)
    return "".join(c for c in stem if not c.isdigit())

# ----------------------------
# LOAD
# ----------------------------

rows = []

for condition, csv_path in TRIAL_CSVS.items():

    with open(CONV1_FILES[condition]) as f:
        emb_raw = json.load(f)

    emb = {clean_name(k): np.array(v) for k,v in emb_raw.items()}

    df = pd.read_csv(csv_path)
    knife_df = df[df["object_response"].astype(str).str.lower()=="knife"]

    for _, row in knife_df.iterrows():
        img = row["imagename"]
        key = clean_name(img)

        if key not in emb:
            continue

        true_cat = str(row["category"]).lower() if "category" in df.columns else get_true_category(img)
        is_true = (true_cat == "knife")

        rows.append({
            "condition": condition,
            "is_true_knife": is_true,
            "embedding": emb[key]
        })

kdf = pd.DataFrame(rows)
print("Total samples:", len(kdf))

# ----------------------------
# MATRIX
# ----------------------------

X = np.stack(kdf["embedding"].values)

pca = PCA(n_components=min(50, X.shape[0]-1, X.shape[1]))
X_pca = pca.fit_transform(X)

tsne = TSNE(
    n_components=2,
    perplexity=max(2, min(30, len(X)//4)),
    random_state=42,
    init="pca",
    max_iter=1000
).fit_transform(X_pca)

# ----------------------------
# PLOT
# ----------------------------

colors = {
    "silhouette":"#2196F3",
    "edges":"#FF2600",
    "outline":"#27B05E",
    "noise_filled":"#D5E91E"
}

fig, ax = plt.subplots(figsize=(12,9))

for i, row in kdf.iterrows():
    ax.scatter(
        tsne[i,0],
        tsne[i,1],
        c=colors[row["condition"]],
        marker="X" if row["is_true_knife"] else "o",
        s=60,
        alpha=0.8
    )

ax.set_title("Conv1 t-SNE Knife Embeddings")
ax.set_xlabel("t-SNE 1")
ax.set_ylabel("t-SNE 2")

handles = [mpatches.Patch(color=colors[c], label=c) for c in colors]
ax.legend(handles=handles)

plt.tight_layout()

# ----------------------------
# SAVE (IMPORTANT FIX)
# ----------------------------

save_path = f"{BASE_DIR}/results/plots/conv1_knife_tsne.png"
os.makedirs(os.path.dirname(save_path), exist_ok=True)

fig.savefig(save_path, dpi=300, bbox_inches="tight")

print("Saved to:", save_path)

plt.show()
plt.close(fig)
```

```python
import os
import json
import torch
import numpy as np
from PIL import Image
from torchvision import models, transforms
from tqdm import tqdm

# --------------------------------------------------
# DEVICE
# --------------------------------------------------

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Using:", device)

# --------------------------------------------------
# RESNET50 AVGPOOL FEATURES (2048-D)
# --------------------------------------------------

model = models.resnet50(weights="DEFAULT")

model = torch.nn.Sequential(

    *(list(model.children())[:-1])

)

model.eval()

model.to(device)

# --------------------------------------------------
# PREPROCESS
# --------------------------------------------------

transform = transforms.Compose([

    transforms.Resize((224,224)),

    transforms.ToTensor(),

    transforms.Normalize(

        mean=[0.485,0.456,0.406],

        std=[0.229,0.224,0.225]
    )
])

# --------------------------------------------------
# INPUT / OUTPUT PATHS
# --------------------------------------------------

conditions = {

    "edges": (

        "/content/drive/MyDrive/AI_Mistakes_Error_Framework/stimuli_original_novel/geirhos-edges",

        "/content/drive/MyDrive/AI_Mistakes_Error_Framework/edges_embeddings.json"

    ),

    "outline": (

        "/content/drive/MyDrive/AI_Mistakes_Error_Framework/stimuli_original_novel/novel_outlines",

        "/content/drive/MyDrive/AI_Mistakes_Error_Framework/outline_embeddings.json"

    ),

    "noise": (

        "/content/drive/MyDrive/AI_Mistakes_Error_Framework/stimuli_original_novel/novel_noise_filled",

        "/content/drive/MyDrive/AI_Mistakes_Error_Framework/noise_embeddings.json"

    )
}

# --------------------------------------------------
# EXTRACT
# --------------------------------------------------

for cond,(folder,save_file) in conditions.items():

    print("\nProcessing:", cond)

    embeddings = {}

    image_paths = []

    for root,dirs,files in os.walk(folder):

        for f in files:

            if f.lower().endswith(

                (".png",".jpg",".jpeg")

            ):

                image_paths.append(

                    os.path.join(root,f)
                )

    print("Images:",len(image_paths))

    for path in tqdm(image_paths):

        try:

            img = Image.open(

                path

            ).convert("RGB")

            x = transform(img)

            x = x.unsqueeze(0)

            x = x.to(device)

            with torch.no_grad():

                feat = model(x)

            feat = feat.squeeze()

            feat = feat.cpu().numpy()

            rel = os.path.relpath(

                path,

                folder
            )

            embeddings[rel] = feat.tolist()

        except:

            continue

    with open(save_file,"w") as f:

        json.dump(

            embeddings,

            f
        )

    print(

        "saved:",

        save_file,

        "n =",

        len(embeddings)
    )

print("\nDONE")
```

```python
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
import os

# =====================================================
# FILES
# =====================================================

AVGPOOL_FILES = {

    "silhouette":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/silhouette_embeddings.json",

    "edges":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/edges_embeddings.json",

    "outline":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/outline_embeddings.json",

    "noise_filled":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/noise_embeddings.json"
}

TRIAL_CSVS = {

    "silhouette":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/trial_data/black_filled_silhouettes_trial_data.csv",

    "edges":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/trial_data/original_edges_trial_data_FIXED.csv",

    "outline":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/trial_data/black_filled_silhouettes_trial_data.csv",

    "noise_filled":
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/trial_data/noise_filled_silhouettes_trial_data.csv"
}

# =====================================================
# NORMALIZE FILENAMES
# =====================================================

def clean_name(x):

    x = str(x)

    x = os.path.basename(x)

    x = x.lower()

    x = x.replace("_edges","")

    x = x.replace("_hq_outline","")

    x = x.replace("_noise_filled","")

    x = x.replace(".jpeg",".png")

    return x

# =====================================================
# LOAD DATA
# =====================================================

all_embeddings = []
all_conditions = []
knife_truth = []

for condition in AVGPOOL_FILES:

    print("\nProcessing:", condition)

    # embeddings
    with open(AVGPOOL_FILES[condition]) as f:

        emb = json.load(f)

    emb_keys = {clean_name(k):v for k,v in emb.items()}

    print("Embeddings loaded:", len(emb_keys))

    # csv
    df = pd.read_csv(TRIAL_CSVS[condition])

    print("CSV rows:", len(df))

    # inspect columns
    print(df.columns)

    # detect filename column automatically
    possible_cols = [c for c in df.columns
                     if "image" in c.lower()
                     or "file" in c.lower()
                     or "stim" in c.lower()]

    fname_col = possible_cols[0]

    # detect response column automatically
    resp_col = [c for c in df.columns
                if "response" in c.lower()
                or "answer" in c.lower()][0]

    df["clean"] = df[fname_col].apply(clean_name)

    matches = 0

    for _,row in df.iterrows():

        fname = row["clean"]

        if fname not in emb_keys:

            continue

        matches += 1

        response = str(row[resp_col]).lower()

        trueknife = (response == "knife")

        all_embeddings.append(
            emb_keys[fname]
        )

        all_conditions.append(
            condition
        )

        knife_truth.append(
            trueknife
        )

    print("Matched:", matches)

# =====================================================
# VERIFY COUNTS
# =====================================================

final_df = pd.DataFrame({

    "condition":all_conditions,
    "knife":knife_truth
})

print("\nFINAL COUNTS")

print(
    final_df.groupby("condition").size()
)

# =====================================================
# TSNE
# =====================================================

X = np.array(all_embeddings)

print("\nTotal embeddings:", X.shape)

tsne = TSNE(

    n_components=2,
    perplexity=30,
    random_state=42

)

coords = tsne.fit_transform(X)

plot_df = pd.DataFrame({

    "x":coords[:,0],
    "y":coords[:,1],
    "condition":all_conditions,
    "knife":knife_truth
})

# =====================================================
# PLOT
# =====================================================

colors = {

    "silhouette":"tab:blue",
    "edges":"tab:red",
    "outline":"tab:green",
    "noise_filled":"y"
}

plt.figure(figsize=(12,9))

for cond in colors:

    subset = plot_df[
        plot_df.condition==cond
    ]

    falsek = subset[
        subset.knife==False
    ]

    truek = subset[
        subset.knife==True
    ]

    plt.scatter(

        falsek.x,
        falsek.y,

        c=colors[cond],

        alpha=.7,

        label=cond
    )

    plt.scatter(

        truek.x,
        truek.y,

        marker="X",

        edgecolors="black",

        c=colors[cond],

        s=100
    )

plt.legend()

plt.title(
"avgpool embeddings: knife true/false labels across conditions\nCircles = true knife | X (black edge) = false knife"
)

plt.xlabel("t-SNE 1")

plt.ylabel("t-SNE 2")

plt.show()
```

```python
import zipfile
import os
import shutil

zip_path = "/content/gradcam_outputs_16class.zip"

# ---------------------------------------------------------
# Condition → CSV + Grad-CAM folder
# ---------------------------------------------------------

conditions = {
    "original_edges": {
        "csv": "original_edges_trial_data_FIXED.csv",
        "zip_folder": "gradcam_outputs_16class/edges/"
    },

    "black_filled_silhouettes": {
        "csv": "black_filled_silhouettes_trial_data.csv",
        "zip_folder": "gradcam_outputs_16class/filled-silhouettes/"
    },

    "original_cue_conflict": {
        "csv": "original_cue-conflict_trial_data.csv",
        "zip_folder": "gradcam_outputs_16class/style-transfer-preprocessed-512/"
    }
}

# ---------------------------------------------------------
# Output folder
# ---------------------------------------------------------

base_output = "/content/false_knife_gradcams"

os.makedirs(base_output, exist_ok=True)

# ---------------------------------------------------------
# Open ZIP
# ---------------------------------------------------------

with zipfile.ZipFile(zip_path, "r") as z:

    zip_files = z.namelist()

    # -----------------------------------------------------
    # Process each condition
    # -----------------------------------------------------

    for condition, info in conditions.items():

        print("\n" + "="*70)
        print(condition.upper())
        print("="*70)

        # Load corresponding CSV
        df = data[info["csv"]]

        # Identify false-knife cases
        false_knife = df[
            (df["category"] != "knife") &
            (df["object_response"] == "knife")
        ]

        # Create condition folder
        output_folder = os.path.join(
            base_output,
            condition
        )

        os.makedirs(output_folder, exist_ok=True)

        found = 0
        missing = []

        # -------------------------------------------------
        # Find corresponding Grad-CAM for every false knife
        # -------------------------------------------------

        for _, row in false_knife.iterrows():

            image_name = row["imagename"]

            # airplane1.png
            # →
            # airplane1_gradcam_knife.jpg
            base_name = os.path.splitext(image_name)[0]

            expected_name = base_name + "_gradcam_knife.jpg"

            matches = [
                f for f in zip_files
                if f.startswith(info["zip_folder"])
                and f.endswith("/" + expected_name)
            ]

            if matches:

                source_file = matches[0]

                output_file = os.path.join(
                    output_folder,
                    expected_name
                )

                with z.open(source_file) as source:
                    with open(output_file, "wb") as target:
                        shutil.copyfileobj(source, target)

                found += 1

            else:
                missing.append(image_name)

        # -------------------------------------------------
        # Summary
        # -------------------------------------------------

        print("False-knife cases:", len(false_knife))
        print("Grad-CAMs extracted:", found)
        print("Missing:", len(missing))

        if missing:
            print("\nMissing files:")
            for x in missing:
                print("  ", x)

print("\n" + "="*70)
print("ALL EXTRACTION COMPLETE")
print("="*70)

print("\nOutput folder:")
print(base_output)
```

```python
from PIL import Image, ImageDraw
import os
import math

folder = "/content/false_knife_gradcams/original_edges"

files = sorted([
    f for f in os.listdir(folder)
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
])

print("Number of Grad-CAMs:", len(files))

# Settings
thumb_width = 180
thumb_height = 180
label_height = 35
columns = 5
rows = math.ceil(len(files) / columns)

sheet_width = columns * thumb_width
sheet_height = rows * (thumb_height + label_height)

# Create contact sheet
sheet = Image.new("RGB", (sheet_width, sheet_height), "white")
draw = ImageDraw.Draw(sheet)

for i, filename in enumerate(files):

    path = os.path.join(folder, filename)

    img = Image.open(path).convert("RGB")
    img.thumbnail((thumb_width - 10, thumb_height - 10))

    x = (i % columns) * thumb_width
    y = (i // columns) * (thumb_height + label_height)

    # Center image
    img_x = x + (thumb_width - img.width) // 2
    img_y = y + (thumb_height - img.height) // 2

    sheet.paste(img, (img_x, img_y))

    # Label
    label = filename.replace("_gradcam_knife.jpg", "")

    draw.text(
        (x + 5, y + thumb_height),
        label,
        fill="black"
    )

# Save
output = "/content/original_edges_false_knife_contact_sheet.jpg"
sheet.save(output, quality=95)

print("Saved:", output)
```

```python
import os
import pandas as pd
import requests
from pathlib import Path

# ============================================================
# 1. PATHS
# ============================================================

OUTPUT = Path("/content/false_knife_analysis")
OUTPUT.mkdir(exist_ok=True)

# Your extracted Grad-CAMs
GRADCAM_ROOT = Path("/content/false_knife_gradcams")

# ============================================================
# 2. CSV FILES
# ============================================================

files = {
    "edges": "/content/original_edges_trial_data_FIXED.csv",
    "silhouettes": "/content/black_filled_silhouettes_trial_data.csv",
    "cue_conflict": "/content/original_cue-conflict_trial_data.csv"
}

# ============================================================
# 3. GITHUB BASE
# ============================================================

GITHUB_RAW = (
    "https://raw.githubusercontent.com/"
    "Darshini-CogSci/AI_Mistakes_Error_Framework/main/"
)

# Repository locations
REPO_PATHS = {
    "edges":
        "stimuli_original_novel/geirhos-edges",

    "silhouettes":
        "stimuli_original_novel/geirhos-silhouettes",

    "cue_conflict":
        "stimuli_original_novel/geirhos-cue-conflict"
}

# ============================================================
# 4. IDENTIFY FALSE-KNIFE CASES
# ============================================================

def get_false_knives(csv_path):

    df = pd.read_csv(csv_path)

    # A false knife = actual category is NOT knife
    # but model responded "knife"
    false_knives = df[
        (df["category"] != "knife") &
        (df["object_response"] == "knife")
    ].copy()

    return false_knives


all_cases = {}

for condition, csv_path in files.items():

    df = get_false_knives(csv_path)
    all_cases[condition] = df

    print(
        f"{condition:15s}: "
        f"{len(df)} false-knife cases"
    )

print("\nTOTAL:",
      sum(len(x) for x in all_cases.values()))


# ============================================================
# 5. FIND GRAD-CAM FILE
# ============================================================

def find_gradcam(condition, imagename):

    # Expected filename:
    # airplane1.png
    # -> airplane1_gradcam_knife.jpg

    stem = Path(imagename).stem
    target = stem + "_gradcam_knife.jpg"

    matches = list(
        GRADCAM_ROOT.rglob(target)
    )

    if len(matches) == 0:
        return None

    # Normally there should only be one
    return matches[0]


# ============================================================
# 6. GITHUB IMAGE NAME
# ============================================================

def github_filename(condition, imagename):

    stem = Path(imagename).stem

    if condition == "edges":

        # airplane1.png
        # ->
        # airplane1_edges.png

        return stem + "_edges.png"

    elif condition == "silhouettes":

        # bear3.png
        # ->
        # bear3.png

        return imagename

    elif condition == "cue_conflict":

        # bear4-knife1.png
        # ->
        # bear4-knife1.png

        return imagename


# ============================================================
# 7. DOWNLOAD + ORGANIZE
# ============================================================

summary = []

for condition, df in all_cases.items():

    print("\n" + "="*70)
    print(condition.upper())
    print("="*70)

    for _, row in df.iterrows():

        imagename = row["imagename"]
        category = row["category"]

        # --------------------------------------------
        # Create folder:
        # condition / actual_category
        # --------------------------------------------

        out_dir = OUTPUT / condition / category
        out_dir.mkdir(
            parents=True,
            exist_ok=True
        )

        # --------------------------------------------
        # Original stimulus filename
        # --------------------------------------------

        github_name = github_filename(
            condition,
            imagename
        )

        # --------------------------------------------
        # GitHub URL
        # --------------------------------------------

        url = (
            GITHUB_RAW
            + REPO_PATHS[condition]
            + "/"
            + category
            + "/"
            + github_name
        )

        # --------------------------------------------
        # Download original image
        # --------------------------------------------

        original_path = out_dir / github_name

        try:

            r = requests.get(
                url,
                timeout=20
            )

            if r.status_code == 200:

                with open(
                    original_path,
                    "wb"
                ) as f:
                    f.write(r.content)

                original_status = "OK"

            else:

                original_status = (
                    f"HTTP {r.status_code}"
                )

        except Exception as e:

            original_status = str(e)

        # --------------------------------------------
        # Find Grad-CAM
        # --------------------------------------------

        gradcam_source = find_gradcam(
            condition,
            imagename
        )

        if gradcam_source is not None:

            gradcam_path = (
                out_dir /
                (Path(imagename).stem +
                 "_gradcam_knife.jpg")
            )

            import shutil

            shutil.copy2(
                gradcam_source,
                gradcam_path
            )

            gradcam_status = "OK"

        else:

            gradcam_status = "NOT FOUND"

        # --------------------------------------------
        # Save record
        # --------------------------------------------

        summary.append({
            "condition": condition,
            "category": category,
            "imagename": imagename,
            "model_response": row["object_response"],
            "original_status": original_status,
            "gradcam_status": gradcam_status,
            "original_path": str(original_path),
            "gradcam_source": (
                str(gradcam_source)
                if gradcam_source
                else None
            )
        })


# ============================================================
# 8. SAVE MASTER TABLE
# ============================================================

summary_df = pd.DataFrame(summary)

summary_df.to_csv(
    OUTPUT / "false_knife_master_table.csv",
    index=False
)

# ============================================================
# 9. FINAL CHECK
# ============================================================

print("\n\n" + "="*70)
print("FINAL CHECK")
print("="*70)

print(
    summary_df
    .groupby("condition")
    .agg(
        cases=("imagename", "count"),
        originals_ok=(
            "original_status",
            lambda x: (x == "OK").sum()
        ),
        gradcams_ok=(
            "gradcam_status",
            lambda x: (x == "OK").sum()
        )
    )
)

print("\nTOTAL CASES:", len(summary_df))

print(
    "Original images downloaded:",
    (summary_df["original_status"] == "OK").sum()
)

print(
    "Grad-CAMs found:",
    (summary_df["gradcam_status"] == "OK").sum()
)

print(
    "\nOutput folder:",
    OUTPUT
)
```

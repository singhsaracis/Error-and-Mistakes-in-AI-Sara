```python
# ============================================================
# STEP 2 — BUILD EDGES PROBE DATASET
# Match CSV labels to Layer 2 / Layer 3 embeddings
# ============================================================

import json
import os
import numpy as np
import pandas as pd

BASE = "/content/drive/MyDrive/AI_Mistakes_Error_Framework"

# ------------------------------------------------------------
# Load embeddings
# ------------------------------------------------------------

with open(
    os.path.join(BASE, "layer2_edges_embeddings.json"), "r"
) as f:
    layer2 = json.load(f)

with open(
    os.path.join(BASE, "layer3_edges_embeddings.json"), "r"
) as f:
    layer3 = json.load(f)

# ------------------------------------------------------------
# Load original Edges trial data
# ------------------------------------------------------------

CSV = "/content/drive/MyDrive/original_edges_trial_data_FIXED.csv"

df = pd.read_csv(CSV).copy()

# ------------------------------------------------------------
# Create embedding filename
# CSV: airplane1.png
# Embedding: airplane1_edges.png
# ------------------------------------------------------------

df["embedding_key"] = df["imagename"].apply(
    lambda x: os.path.splitext(x)[0] + "_edges.png"
)

# ------------------------------------------------------------
# Check matching
# ------------------------------------------------------------

df["layer2_found"] = df["embedding_key"].isin(layer2)
df["layer3_found"] = df["embedding_key"].isin(layer3)

print("=" * 70)
print("EDGES EMBEDDING MATCHING")
print("=" * 70)

print("Total CSV trials:", len(df))

print(
    "Layer 2 matches:",
    df["layer2_found"].sum(),
    "/",
    len(df)
)

print(
    "Layer 3 matches:",
    df["layer3_found"].sum(),
    "/",
    len(df)
)

# ------------------------------------------------------------
# Show missing matches if any
# ------------------------------------------------------------

missing_l2 = df.loc[
    ~df["layer2_found"],
    ["imagename", "embedding_key"]
]

missing_l3 = df.loc[
    ~df["layer3_found"],
    ["imagename", "embedding_key"]
]

if len(missing_l2) > 0:
    print("\nMISSING LAYER 2 MATCHES:")
    print(missing_l2)

if len(missing_l3) > 0:
    print("\nMISSING LAYER 3 MATCHES:")
    print(missing_l3)

# ------------------------------------------------------------
# Create labels
#
# Ground truth:
# knife = 1
# everything else = 0
# ------------------------------------------------------------

df["true_label"] = (
    df["category"]
    .astype(str)
    .str.lower()
    .eq("knife")
    .astype(int)
)

# ------------------------------------------------------------
# ResNet's original response
#
# knife = 1
# non-knife = 0
# ------------------------------------------------------------

df["model_label"] = (
    df["object_response"]
    .astype(str)
    .str.lower()
    .eq("knife")
    .astype(int)
)

# ------------------------------------------------------------
# Identify model errors
# ------------------------------------------------------------

df["model_correct"] = (
    df["true_label"] == df["model_label"]
)

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("LABEL SUMMARY")
print("=" * 70)

print("\nTrue labels:")
print(
    df["true_label"]
    .value_counts()
    .sort_index()
    .rename(
        index={
            0: "Non-knife",
            1: "Knife"
        }
    )
)

print("\nResNet predictions:")
print(
    df["model_label"]
    .value_counts()
    .sort_index()
    .rename(
        index={
            0: "Non-knife",
            1: "Knife"
        }
    )
)

print("\nResNet errors:", (~df["model_correct"]).sum())
print("ResNet correct:", df["model_correct"].sum())

# ------------------------------------------------------------
# Check the known false-knife count
# ------------------------------------------------------------

false_knives = (
    (df["true_label"] == 0) &
    (df["model_label"] == 1)
)

print(
    "\nFalse-knife cases:",
    false_knives.sum()
)

# ------------------------------------------------------------
# Build activation matrices
# ------------------------------------------------------------

valid = df[
    df["layer2_found"] &
    df["layer3_found"]
].copy()

X_layer2 = np.array([
    layer2[key]
    for key in valid["embedding_key"]
])

X_layer3 = np.array([
    layer3[key]
    for key in valid["embedding_key"]
])

y = valid["true_label"].to_numpy()

model_predictions = valid["model_label"].to_numpy()

model_correct = valid["model_correct"].to_numpy()

print("\n" + "=" * 70)
print("FINAL PROBE MATRICES")
print("=" * 70)

print("Valid images:", len(valid))

print("Layer 2 shape:", X_layer2.shape)
print("Layer 3 shape:", X_layer3.shape)

print("Labels shape:", y.shape)

print("\nDataset ready for probing.")
```

```python
# ============================================================
# STEP 3B — DIRECT LINEAR PROBE
# EDGES × LAYER 2 + LAYER 3 + AVGPOOL
# ============================================================

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    confusion_matrix,
    classification_report
)

import numpy as np
import pandas as pd
import os

print("=" * 70)
print("EDGES — DIRECT LINEAR PROBE")
print("LAYER 2 + LAYER 3 + AVGPOOL")
print("=" * 70)


# ------------------------------------------------------------
# 1. PREPARE AVGPOOL MATCHING
# ------------------------------------------------------------

df_edges["avgpool_key"] = df_edges["imagename"].apply(
    lambda x: (
        os.path.splitext(x)[0]
        .split("/")[0]
        + "/"
        + os.path.splitext(x)[0]
        + "_edges.png"
    )
)

avgpool_matches = []

for key in df_edges["avgpool_key"]:

    if key in avgpool_data:
        avgpool_matches.append(
            np.array(avgpool_data[key])
        )
    else:
        avgpool_matches.append(None)


print("\nAvgPool matches:",
      sum(x is not None for x in avgpool_matches),
      "/",
      len(avgpool_matches))


# ------------------------------------------------------------
# 2. BUILD AVGPOOL MATRIX
# ------------------------------------------------------------

valid_avgpool = [
    x is not None
    for x in avgpool_matches
]

if not all(valid_avgpool):

    print("\n❌ Some AvgPool embeddings did not match.")

else:

    X_avgpool = np.vstack(avgpool_matches)

    print("AvgPool matrix:",
          X_avgpool.shape)


# ------------------------------------------------------------
# 3. COMMON TRAIN / TEST SPLIT
# ------------------------------------------------------------

indices = np.arange(len(y))

train_idx, test_idx = train_test_split(
    indices,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("\nTraining images:", len(train_idx))
print("Test images:", len(test_idx))

print("\nTraining labels:")
print("  Knife:", y[train_idx].sum())
print("  Non-knife:",
      (y[train_idx] == 0).sum())

print("\nTest labels:")
print("  Knife:", y[test_idx].sum())
print("  Non-knife:",
      (y[test_idx] == 0).sum())


# ------------------------------------------------------------
# 4. FUNCTION FOR ONE PROBE
# ------------------------------------------------------------

def run_probe(X, layer_name):

    X_train = X[train_idx]
    X_test = X[test_idx]

    y_train = y[train_idx]
    y_test = y[test_idx]

    correct_test = model_correct[test_idx]

    probe = LogisticRegression(
        class_weight="balanced",
        max_iter=5000,
        random_state=42
    )

    probe.fit(
        X_train,
        y_train
    )

    probe_pred = probe.predict(X_test)

    overall_accuracy = accuracy_score(
        y_test,
        probe_pred
    )

    balanced_accuracy = balanced_accuracy_score(
        y_test,
        probe_pred
    )

    print("\n")
    print("=" * 70)
    print(layer_name)
    print("=" * 70)

    print(
        f"Overall probe accuracy: "
        f"{overall_accuracy * 100:.2f}%"
    )

    print(
        f"Balanced accuracy: "
        f"{balanced_accuracy * 100:.2f}%"
    )

    print("\nConfusion matrix:")
    print(
        confusion_matrix(
            y_test,
            probe_pred
        )
    )

    # --------------------------------------------------------
    # RESNET ERROR SUBSET
    # --------------------------------------------------------

    resnet_error_test = ~correct_test

    print("\nResNet errors in test set:",
          resnet_error_test.sum())

    if resnet_error_test.sum() > 0:

        error_y = y_test[
            resnet_error_test
        ]

        error_probe_pred = probe_pred[
            resnet_error_test
        ]

        error_accuracy = accuracy_score(
            error_y,
            error_probe_pred
        )

        error_balanced_accuracy = (
            balanced_accuracy_score(
                error_y,
                error_probe_pred
            )
            if len(np.unique(error_y)) > 1
            else np.nan
        )

        print(
            f"Probe accuracy on ResNet errors: "
            f"{error_accuracy * 100:.2f}%"
        )

        if not np.isnan(error_balanced_accuracy):

            print(
                f"Probe balanced accuracy on "
                f"ResNet errors: "
                f"{error_balanced_accuracy * 100:.2f}%"
            )

        print("\nError subset:")
        print("  Knife:",
              error_y.sum())

        print("  Non-knife:",
              (error_y == 0).sum())

        print("\nError-case confusion matrix:")
        print(
            confusion_matrix(
                error_y,
                error_probe_pred
            )
        )

    else:

        error_accuracy = np.nan

    return {
        "layer": layer_name,
        "overall_accuracy": overall_accuracy,
        "balanced_accuracy": balanced_accuracy,
        "resnet_error_n": int(
            resnet_error_test.sum()
        ),
        "error_accuracy": error_accuracy
    }


# ------------------------------------------------------------
# 5. RUN ALL THREE
# ------------------------------------------------------------

results_edges = []

results_edges.append(
    run_probe(
        X_layer2,
        "EDGES — LAYER 2"
    )
)

results_edges.append(
    run_probe(
        X_layer3,
        "EDGES — LAYER 3"
    )
)

results_edges.append(
    run_probe(
        X_avgpool,
        "EDGES — AVGPOOL"
    )
)


# ------------------------------------------------------------
# 6. SUMMARY TABLE
# ------------------------------------------------------------

results_edges_df = pd.DataFrame(
    results_edges
)

print("\n")
print("=" * 70)
print("EDGES — FINAL PROBE SUMMARY")
print("=" * 70)

print(
    results_edges_df.to_string(
        index=False
    )
)
```

```python
# ============================================================
# STEP 3B — EDGES DIRECT LINEAR PROBE
# LAYER 2 + LAYER 3 + AVGPOOL
# ============================================================

import os
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    confusion_matrix
)

print("=" * 70)
print("EDGES — DIRECT LINEAR PROBE")
print("LAYER 2 + LAYER 3 + AVGPOOL")
print("=" * 70)


# ============================================================
# 1. CHECK EXISTING DATA
# ============================================================

print("\nDATA MATRICES")

print("Layer 2:", X_layer2.shape)
print("Layer 3:", X_layer3.shape)


# ============================================================
# 2. MATCH AVGPOOL EMBEDDINGS
# ============================================================

avgpool_matches = []

for imagename in df_edges["imagename"]:

    # Example:
    # airplane1.png
    #
    # becomes:
    # airplane/airplane1_edges.png

    stem = os.path.splitext(imagename)[0]

    category = stem.rstrip("0123456789")

    key = (
        category
        + "/"
        + stem
        + "_edges.png"
    )

    if key in avgpool_data:
        avgpool_matches.append(
            np.array(avgpool_data[key])
        )
    else:
        avgpool_matches.append(None)


print("\nAvgPool matches:",
      sum(x is not None for x in avgpool_matches),
      "/",
      len(avgpool_matches))


if not all(x is not None for x in avgpool_matches):

    print("\n❌ AvgPool matching failed.")
    raise ValueError("Missing AvgPool embeddings.")

X_avgpool = np.vstack(avgpool_matches)

print("AvgPool:", X_avgpool.shape)


# ============================================================
# 3. VERIFY ALL MATRICES HAVE SAME NUMBER OF IMAGES
# ============================================================

assert len(X_layer2) == len(X_layer3)
assert len(X_layer2) == len(X_avgpool)
assert len(X_layer2) == len(y)

print("\nAll representations contain:",
      len(y),
      "images")


# ============================================================
# 4. TRAIN / TEST SPLIT
# ============================================================

indices = np.arange(len(y))

train_idx, test_idx = train_test_split(
    indices,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("\n" + "=" * 70)
print("TRAIN / TEST SPLIT")
print("=" * 70)

print("Training images:", len(train_idx))
print("Test images:", len(test_idx))

print("\nTraining labels:")
print("  Knife:", int(y[train_idx].sum()))
print("  Non-knife:",
      int((y[train_idx] == 0).sum()))

print("\nTest labels:")
print("  Knife:", int(y[test_idx].sum()))
print("  Non-knife:",
      int((y[test_idx] == 0).sum()))


# ============================================================
# 5. FUNCTION TO RUN ONE LINEAR PROBE
# ============================================================

def run_linear_probe(X, layer_name):

    X_train = X[train_idx]
    X_test = X[test_idx]

    y_train = y[train_idx]
    y_test = y[test_idx]

    correct_test = model_correct[test_idx]

    # --------------------------------------------------------
    # Train probe
    # --------------------------------------------------------

    probe = LogisticRegression(
        class_weight="balanced",
        max_iter=5000,
        random_state=42
    )

    probe.fit(
        X_train,
        y_train
    )

    # --------------------------------------------------------
    # Predictions
    # --------------------------------------------------------

    probe_pred = probe.predict(
        X_test
    )

    # --------------------------------------------------------
    # Overall performance
    # --------------------------------------------------------

    overall_accuracy = accuracy_score(
        y_test,
        probe_pred
    )

    balanced_accuracy = balanced_accuracy_score(
        y_test,
        probe_pred
    )

    print("\n")
    print("=" * 70)
    print(layer_name)
    print("=" * 70)

    print(
        f"Overall probe accuracy: "
        f"{overall_accuracy * 100:.2f}%"
    )

    print(
        f"Balanced accuracy: "
        f"{balanced_accuracy * 100:.2f}%"
    )

    print("\nConfusion matrix:")

    print(
        confusion_matrix(
            y_test,
            probe_pred
        )
    )

    # --------------------------------------------------------
    # ResNet error subset
    # --------------------------------------------------------

    resnet_error_test = ~correct_test

    n_errors = int(
        resnet_error_test.sum()
    )

    print("\nResNet errors in test set:",
          n_errors)

    error_accuracy = np.nan
    error_balanced_accuracy = np.nan
    error_n = 0
    error_knives = 0
    error_nonknives = 0

    if n_errors > 0:

        error_y = y_test[
            resnet_error_test
        ]

        error_probe_pred = probe_pred[
            resnet_error_test
        ]

        error_accuracy = accuracy_score(
            error_y,
            error_probe_pred
        )

        error_n = len(error_y)

        error_knives = int(
            error_y.sum()
        )

        error_nonknives = int(
            (error_y == 0).sum()
        )

        print(
            f"Probe accuracy on ResNet errors: "
            f"{error_accuracy * 100:.2f}%"
        )

        print("\nError subset:")
        print("  Knife:", error_knives)
        print("  Non-knife:", error_nonknives)

        print("\nError-case confusion matrix:")

        print(
            confusion_matrix(
                error_y,
                error_probe_pred
            )
        )

        # Balanced accuracy only if both classes
        # are present in the error subset

        if len(np.unique(error_y)) == 2:

            error_balanced_accuracy = (
                balanced_accuracy_score(
                    error_y,
                    error_probe_pred
                )
            )

    return {
        "Representation": layer_name,
        "Overall Accuracy": overall_accuracy,
        "Balanced Accuracy": balanced_accuracy,
        "ResNet Errors (Test)": n_errors,
        "Error Accuracy": error_accuracy,
        "Error Balanced Accuracy": error_balanced_accuracy,
        "Error Knives": error_knives,
        "Error Non-knives": error_nonknives
    }


# ============================================================
# 6. RUN LAYER 2
# ============================================================

result_layer2 = run_linear_probe(
    X_layer2,
    "EDGES — LAYER 2"
)


# ============================================================
# 7. RUN LAYER 3
# ============================================================

result_layer3 = run_linear_probe(
    X_layer3,
    "EDGES — LAYER 3"
)


# ============================================================
# 8. RUN AVGPOOL
# ============================================================

result_avgpool = run_linear_probe(
    X_avgpool,
    "EDGES — AVGPOOL"
)


# ============================================================
# 9. FINAL COMPARISON TABLE
# ============================================================

results_edges = pd.DataFrame([
    result_layer2,
    result_layer3,
    result_avgpool
])

print("\n\n")
print("=" * 70)
print("EDGES — FINAL THREE-REPRESENTATION COMPARISON")
print("=" * 70)

print(
    results_edges.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# ============================================================
# 10. PERCENTAGE VERSION
# ============================================================

print("\n" + "=" * 70)
print("EDGES — RESULTS IN PERCENTAGES")
print("=" * 70)

for _, row in results_edges.iterrows():

    print(
        f"\n{row['Representation']}"
    )

    print(
        f"  Overall accuracy: "
        f"{row['Overall Accuracy'] * 100:.2f}%"
    )

    print(
        f"  Balanced accuracy: "
        f"{row['Balanced Accuracy'] * 100:.2f}%"
    )

    if not np.isnan(row["Error Accuracy"]):

        print(
            f"  ResNet-error accuracy: "
            f"{row['Error Accuracy'] * 100:.2f}%"
        )

        print(
            f"  ResNet-error cases: "
            f"{int(row['ResNet Errors (Test)'])}"
        )
```

```python
# ============================================================
# STEP 3C — SILHOUETTES DIRECT LINEAR PROBE
# LAYER 2 + LAYER 3 + AVGPOOL
# ============================================================

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    confusion_matrix
)

import numpy as np
import pandas as pd


print("=" * 70)
print("SILHOUETTES — DIRECT LINEAR PROBE")
print("LAYER 2 + LAYER 3 + AVGPOOL")
print("=" * 70)


# ============================================================
# 1. VERIFY MATRICES
# ============================================================

print("\nDATA MATRICES")

print("Layer 2:", X_layer2_sil.shape)
print("Layer 3:", X_layer3_sil.shape)
print("AvgPool:", X_avgpool_sil.shape)


assert len(X_layer2_sil) == len(X_layer3_sil)
assert len(X_layer2_sil) == len(X_avgpool_sil)
assert len(X_layer2_sil) == len(y_sil)

print(
    "All representations contain:",
    len(y_sil),
    "images"
)


# ============================================================
# 2. TRAIN / TEST SPLIT
# ============================================================

indices_sil = np.arange(len(y_sil))

train_idx_sil, test_idx_sil = train_test_split(
    indices_sil,
    test_size=0.30,
    random_state=42,
    stratify=y_sil
)

print("\n" + "=" * 70)
print("TRAIN / TEST SPLIT")
print("=" * 70)

print(
    "Training images:",
    len(train_idx_sil)
)

print(
    "Test images:",
    len(test_idx_sil)
)

print("\nTraining labels:")
print(
    "  Knife:",
    int(y_sil[train_idx_sil].sum())
)

print(
    "  Non-knife:",
    int((y_sil[train_idx_sil] == 0).sum())
)

print("\nTest labels:")
print(
    "  Knife:",
    int(y_sil[test_idx_sil].sum())
)

print(
    "  Non-knife:",
    int((y_sil[test_idx_sil] == 0).sum())
)


# ============================================================
# 3. FUNCTION FOR ONE PROBE
# ============================================================

def run_silhouette_probe(
    X,
    layer_name
):

    X_train = X[train_idx_sil]
    X_test = X[test_idx_sil]

    y_train = y_sil[train_idx_sil]
    y_test = y_sil[test_idx_sil]

    correct_test = model_correct_sil[
        test_idx_sil
    ]

    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    probe = LogisticRegression(
        class_weight="balanced",
        max_iter=5000,
        random_state=42
    )

    probe.fit(
        X_train,
        y_train
    )

    # --------------------------------------------------------
    # PREDICT
    # --------------------------------------------------------

    probe_pred = probe.predict(
        X_test
    )

    # --------------------------------------------------------
    # OVERALL PERFORMANCE
    # --------------------------------------------------------

    overall_accuracy = accuracy_score(
        y_test,
        probe_pred
    )

    balanced_accuracy = balanced_accuracy_score(
        y_test,
        probe_pred
    )

    print("\n")
    print("=" * 70)
    print(layer_name)
    print("=" * 70)

    print(
        f"Overall probe accuracy: "
        f"{overall_accuracy * 100:.2f}%"
    )

    print(
        f"Balanced accuracy: "
        f"{balanced_accuracy * 100:.2f}%"
    )

    print("\nConfusion matrix:")

    print(
        confusion_matrix(
            y_test,
            probe_pred
        )
    )

    # --------------------------------------------------------
    # RESNET ERROR SUBSET
    # --------------------------------------------------------

    resnet_error_test = ~correct_test

    n_errors = int(
        resnet_error_test.sum()
    )

    print(
        "\nResNet errors in test set:",
        n_errors
    )

    error_accuracy = np.nan
    error_balanced_accuracy = np.nan

    error_knives = 0
    error_nonknives = 0

    if n_errors > 0:

        error_y = y_test[
            resnet_error_test
        ]

        error_probe_pred = probe_pred[
            resnet_error_test
        ]

        error_accuracy = accuracy_score(
            error_y,
            error_probe_pred
        )

        error_knives = int(
            error_y.sum()
        )

        error_nonknives = int(
            (error_y == 0).sum()
        )

        print(
            f"Probe accuracy on ResNet errors: "
            f"{error_accuracy * 100:.2f}%"
        )

        print("\nError subset:")
        print(
            "  Knife:",
            error_knives
        )

        print(
            "  Non-knife:",
            error_nonknives
        )

        print("\nError-case confusion matrix:")

        print(
            confusion_matrix(
                error_y,
                error_probe_pred
            )
        )

        if len(np.unique(error_y)) == 2:

            error_balanced_accuracy = (
                balanced_accuracy_score(
                    error_y,
                    error_probe_pred
                )
            )

            print(
                f"\nError balanced accuracy: "
                f"{error_balanced_accuracy * 100:.2f}%"
            )

    return {
        "Representation": layer_name,
        "Overall Accuracy": overall_accuracy,
        "Balanced Accuracy": balanced_accuracy,
        "ResNet Errors (Test)": n_errors,
        "Error Accuracy": error_accuracy,
        "Error Balanced Accuracy": error_balanced_accuracy,
        "Error Knives": error_knives,
        "Error Non-knives": error_nonknives
    }


# ============================================================
# 4. RUN LAYER 2
# ============================================================

result_sil_l2 = run_silhouette_probe(
    X_layer2_sil,
    "SILHOUETTES — LAYER 2"
)


# ============================================================
# 5. RUN LAYER 3
# ============================================================

result_sil_l3 = run_silhouette_probe(
    X_layer3_sil,
    "SILHOUETTES — LAYER 3"
)


# ============================================================
# 6. RUN AVGPOOL
# ============================================================

result_sil_avg = run_silhouette_probe(
    X_avgpool_sil,
    "SILHOUETTES — AVGPOOL"
)


# ============================================================
# 7. FINAL COMPARISON TABLE
# ============================================================

results_sil = pd.DataFrame([
    result_sil_l2,
    result_sil_l3,
    result_sil_avg
])

print("\n\n")
print("=" * 70)
print("SILHOUETTES — FINAL THREE-REPRESENTATION COMPARISON")
print("=" * 70)

print(
    results_sil.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# ============================================================
# 8. PERCENTAGE SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("SILHOUETTES — RESULTS IN PERCENTAGES")
print("=" * 70)

for _, row in results_sil.iterrows():

    print(
        f"\n{row['Representation']}"
    )

    print(
        f"  Overall accuracy: "
        f"{row['Overall Accuracy'] * 100:.2f}%"
    )

    print(
        f"  Balanced accuracy: "
        f"{row['Balanced Accuracy'] * 100:.2f}%"
    )

    if not np.isnan(row["Error Accuracy"]):

        print(
            f"  ResNet-error accuracy: "
            f"{row['Error Accuracy'] * 100:.2f}%"
        )

        print(
            f"  ResNet-error cases: "
            f"{int(row['ResNet Errors (Test)'])}"
        )
```

```python
# ============================================================
# CUE-CONFLICT EMBEDDING EXTRACTION
# Layer 2 | Layer 3 | AvgPool
# ============================================================

import os
import json
import torch
import torch.nn as nn
import pandas as pd
import numpy as np
from PIL import Image
from torchvision import models, transforms

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

csv_path = "/content/drive/MyDrive/AI_Mistakes_Error_Framework/trial_data/original_cue-conflict_trial_data.csv"

checkpoint_path = "/content/drive/MyDrive/false_knife_analysis/resnet50-0676ba61.pth"

# IMPORTANT:
# Change this ONLY if your cue-conflict image folder is elsewhere.
# We will first try to infer image paths from the CSV.
# ------------------------------------------------------------

df_cue = pd.read_csv(csv_path)

print("Rows:", len(df_cue))
print("Unique images:", df_cue["imagename"].nunique())

# ------------------------------------------------------------
# LOAD MODEL
# ------------------------------------------------------------

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

model = models.resnet50(weights=None)

checkpoint = torch.load(
    checkpoint_path,
    map_location=device
)

if "state_dict" in checkpoint:
    checkpoint = checkpoint["state_dict"]

# Remove possible DataParallel prefix
checkpoint = {
    k.replace("module.", ""): v
    for k, v in checkpoint.items()
}

model.load_state_dict(checkpoint, strict=False)
model = model.to(device)
model.eval()

print("ResNet-50 checkpoint loaded.")

# ------------------------------------------------------------
# IMAGE PREPROCESSING
# Same ImageNet preprocessing used previously
# ------------------------------------------------------------

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# ------------------------------------------------------------
# HOOKS
# ------------------------------------------------------------

layer2_outputs = {}
layer3_outputs = {}
avgpool_outputs = {}

def hook_layer2(module, input, output):
    layer2_outputs["x"] = output.detach().cpu()

def hook_layer3(module, input, output):
    layer3_outputs["x"] = output.detach().cpu()

def hook_avgpool(module, input, output):
    avgpool_outputs["x"] = output.detach().cpu()

h2 = model.layer2.register_forward_hook(hook_layer2)
h3 = model.layer3.register_forward_hook(hook_layer3)
hap = model.avgpool.register_forward_hook(hook_avgpool)

# ------------------------------------------------------------
# FIRST: FIND IMAGE ROOT
# ------------------------------------------------------------

candidate_roots = [
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework",
    "/content/drive/MyDrive/false_knife_analysis",
    "/content/drive/MyDrive"
]

def find_image(imagename):
    """
    Search for the exact cue-conflict filename.
    """
    for root in candidate_roots:
        for dirpath, dirnames, filenames in os.walk(root):
            if imagename in filenames:
                return os.path.join(dirpath, imagename)
    return None

# ------------------------------------------------------------
# EXTRACT
# ------------------------------------------------------------

layer2_embeddings = {}
layer3_embeddings = {}
avgpool_embeddings = {}

missing = []

unique_images = df_cue["imagename"].drop_duplicates().tolist()

print("\nStarting extraction...")
print("Images to process:", len(unique_images))

for i, imagename in enumerate(unique_images):

    image_path = find_image(imagename)

    if image_path is None:
        missing.append(imagename)
        continue

    try:
        image = Image.open(image_path).convert("RGB")
        image = transform(image).unsqueeze(0).to(device)

        # Clear old hook outputs
        layer2_outputs.clear()
        layer3_outputs.clear()
        avgpool_outputs.clear()

        with torch.no_grad():
            model(image)

        # Layer 2: [1, 512, H, W]
        l2 = layer2_outputs["x"].mean(dim=(2, 3)).squeeze(0).numpy()

        # Layer 3: [1, 1024, H, W]
        l3 = layer3_outputs["x"].mean(dim=(2, 3)).squeeze(0).numpy()

        # AvgPool: [1, 2048, 1, 1]
        ap = avgpool_outputs["x"].flatten().numpy()

        layer2_embeddings[imagename] = l2.tolist()
        layer3_embeddings[imagename] = l3.tolist()
        avgpool_embeddings[imagename] = ap.tolist()

    except Exception as e:
        print("FAILED:", imagename, "|", str(e))
        missing.append(imagename)

    if (i + 1) % 100 == 0:
        print(f"Processed {i+1}/{len(unique_images)}")

# Remove hooks
h2.remove()
h3.remove()
hap.remove()

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

output_dir = "/content/drive/MyDrive/false_knife_analysis/cue_conflict_embeddings"
os.makedirs(output_dir, exist_ok=True)

layer2_path = os.path.join(
    output_dir,
    "layer2_cue_conflict_embeddings.json"
)

layer3_path = os.path.join(
    output_dir,
    "layer3_cue_conflict_embeddings.json"
)

avgpool_path = os.path.join(
    output_dir,
    "avgpool_cue_conflict_embeddings.json"
)

with open(layer2_path, "w") as f:
    json.dump(layer2_embeddings, f)

with open(layer3_path, "w") as f:
    json.dump(layer3_embeddings, f)

with open(avgpool_path, "w") as f:
    json.dump(avgpool_embeddings, f)

print("\n============================================")
print("EXTRACTION COMPLETE")
print("============================================")

print("Layer 2:", len(layer2_embeddings))
print("Layer 3:", len(layer3_embeddings))
print("AvgPool:", len(avgpool_embeddings))
print("Missing:", len(missing))

print("\nSaved:")
print(layer2_path)
print(layer3_path)
print(avgpool_path)

if missing:
    print("\nFirst missing images:")
    print(missing[:20])
```

```python
# ============================================================
# CUE-CONFLICT EMBEDDING EXTRACTION — OPTIMIZED
# Layer 2 | Layer 3 | AvgPool
# ============================================================

import os
import json
import torch
import pandas as pd
import numpy as np
from PIL import Image
from torchvision import models, transforms

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

csv_path = "/content/drive/MyDrive/AI_Mistakes_Error_Framework/trial_data/original_cue-conflict_trial_data.csv"

image_root = "/content/drive/MyDrive/AI_Mistakes_Error_Framework/stimuli_original_novel/geirhos-cue-conflict"

checkpoint_path = "/content/drive/MyDrive/false_knife_analysis/resnet50-0676ba61.pth"

output_dir = "/content/drive/MyDrive/false_knife_analysis/cue_conflict_embeddings"

os.makedirs(output_dir, exist_ok=True)

# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------

df_cue = pd.read_csv(csv_path)

print("Rows:", len(df_cue))
print("Unique images:", df_cue["imagename"].nunique())

# ------------------------------------------------------------
# LOAD MODEL
# ------------------------------------------------------------

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)

model = models.resnet50(weights=None)

checkpoint = torch.load(
    checkpoint_path,
    map_location=device
)

if "state_dict" in checkpoint:
    checkpoint = checkpoint["state_dict"]

checkpoint = {
    k.replace("module.", ""): v
    for k, v in checkpoint.items()
}

model.load_state_dict(checkpoint, strict=False)

model = model.to(device)
model.eval()

print("ResNet-50 checkpoint loaded.")

# ------------------------------------------------------------
# PREPROCESSING
# ------------------------------------------------------------

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# ------------------------------------------------------------
# HOOKS
# ------------------------------------------------------------

activations = {}

def hook_layer2(module, input, output):
    activations["layer2"] = output.detach()

def hook_layer3(module, input, output):
    activations["layer3"] = output.detach()

def hook_avgpool(module, input, output):
    activations["avgpool"] = output.detach()

h2 = model.layer2.register_forward_hook(hook_layer2)
h3 = model.layer3.register_forward_hook(hook_layer3)
hap = model.avgpool.register_forward_hook(hook_avgpool)

# ------------------------------------------------------------
# EMBEDDING DICTIONARIES
# ------------------------------------------------------------

layer2_embeddings = {}
layer3_embeddings = {}
avgpool_embeddings = {}

missing = []
failed = []

# ------------------------------------------------------------
# PROCESS UNIQUE IMAGES
# ------------------------------------------------------------

unique_df = df_cue.drop_duplicates("imagename").copy()

print("\nStarting extraction...")
print("Images to process:", len(unique_df))

for i, row in enumerate(unique_df.itertuples(index=False), start=1):

    imagename = row.imagename
    category = row.category

    # Direct path — NO os.walk()
    image_path = os.path.join(
        image_root,
        category,
        imagename
    )

    if not os.path.exists(image_path):
        missing.append((imagename, image_path))
        continue

    try:
        image = Image.open(image_path).convert("RGB")
        image_tensor = transform(image).unsqueeze(0).to(device)

        activations.clear()

        with torch.no_grad():
            model(image_tensor)

        # Layer 2: global average pooling over spatial dimensions
        l2 = (
            activations["layer2"]
            .mean(dim=(2, 3))
            .squeeze(0)
            .cpu()
            .numpy()
        )

        # Layer 3: global average pooling over spatial dimensions
        l3 = (
            activations["layer3"]
            .mean(dim=(2, 3))
            .squeeze(0)
            .cpu()
            .numpy()
        )

        # AvgPool: already 2048-dimensional
        ap = (
            activations["avgpool"]
            .flatten()
            .cpu()
            .numpy()
        )

        layer2_embeddings[imagename] = l2.tolist()
        layer3_embeddings[imagename] = l3.tolist()
        avgpool_embeddings[imagename] = ap.tolist()

    except Exception as e:
        failed.append((imagename, str(e)))

    if i % 100 == 0:
        print(f"Processed {i}/{len(unique_df)}")

# ------------------------------------------------------------
# REMOVE HOOKS
# ------------------------------------------------------------

h2.remove()
h3.remove()
hap.remove()

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

layer2_path = os.path.join(
    output_dir,
    "layer2_cue_conflict_embeddings.json"
)

layer3_path = os.path.join(
    output_dir,
    "layer3_cue_conflict_embeddings.json"
)

avgpool_path = os.path.join(
    output_dir,
    "avgpool_cue_conflict_embeddings.json"
)

with open(layer2_path, "w") as f:
    json.dump(layer2_embeddings, f)

with open(layer3_path, "w") as f:
    json.dump(layer3_embeddings, f)

with open(avgpool_path, "w") as f:
    json.dump(avgpool_embeddings, f)

# ------------------------------------------------------------
# FINAL CHECK
# ------------------------------------------------------------

print("\n============================================")
print("EXTRACTION COMPLETE")
print("============================================")

print("Layer 2:", len(layer2_embeddings))
print("Layer 3:", len(layer3_embeddings))
print("AvgPool:", len(avgpool_embeddings))

print("Missing:", len(missing))
print("Failed:", len(failed))

print("\nEmbedding dimensions:")

if layer2_embeddings:
    print("Layer 2:", len(next(iter(layer2_embeddings.values()))))

if layer3_embeddings:
    print("Layer 3:", len(next(iter(layer3_embeddings.values()))))

if avgpool_embeddings:
    print("AvgPool:", len(next(iter(avgpool_embeddings.values()))))

print("\nSaved files:")
print(layer2_path)
print(layer3_path)
print(avgpool_path)

if missing:
    print("\nFirst missing:")
    print(missing[:10])

if failed:
    print("\nFirst failures:")
    print(failed[:10])
```

```python
# ============================================================
# CUE-CONFLICT — DIRECT LINEAR PROBE
# ============================================================

import os
import json
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    confusion_matrix
)

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

csv_path = (
    "/content/drive/MyDrive/"
    "AI_Mistakes_Error_Framework/trial_data/"
    "original_cue-conflict_trial_data.csv"
)

embed_dir = (
    "/content/drive/MyDrive/"
    "false_knife_analysis/cue_conflict_embeddings"
)

# ------------------------------------------------------------
# LOAD CSV
# ------------------------------------------------------------

df = pd.read_csv(csv_path)

# Ground truth: knife = 1, non-knife = 0
df["true_label"] = (
    df["category"].astype(str).str.lower() == "knife"
).astype(int)

# ResNet binary prediction
df["resnet_pred"] = (
    df["object_response"].astype(str).str.lower() == "knife"
).astype(int)

# ------------------------------------------------------------
# LOAD EMBEDDINGS
# ------------------------------------------------------------

with open(
    os.path.join(
        embed_dir,
        "layer2_cue_conflict_embeddings.json"
    ), "r"
) as f:
    layer2_dict = json.load(f)

with open(
    os.path.join(
        embed_dir,
        "layer3_cue_conflict_embeddings.json"
    ), "r"
) as f:
    layer3_dict = json.load(f)

with open(
    os.path.join(
        embed_dir,
        "avgpool_cue_conflict_embeddings.json"
    ), "r"
) as f:
    avgpool_dict = json.load(f)

# ------------------------------------------------------------
# MATCH CSV TO EMBEDDINGS
# ------------------------------------------------------------

df["embedding_key"] = df["imagename"].astype(str)

print("============================================")
print("CUE-CONFLICT DATA CHECK")
print("============================================")

print("CSV rows:", len(df))
print("Layer 2 matches:",
      df["embedding_key"].isin(layer2_dict).sum())
print("Layer 3 matches:",
      df["embedding_key"].isin(layer3_dict).sum())
print("AvgPool matches:",
      df["embedding_key"].isin(avgpool_dict).sum())

# ------------------------------------------------------------
# BUILD MATRICES
# ------------------------------------------------------------

X_layer2 = np.array([
    layer2_dict[k]
    for k in df["embedding_key"]
])

X_layer3 = np.array([
    layer3_dict[k]
    for k in df["embedding_key"]
])

X_avgpool = np.array([
    avgpool_dict[k]
    for k in df["embedding_key"]
])

y = df["true_label"].values

resnet_pred = df["resnet_pred"].values

print("\nMatrix shapes:")
print("Layer 2 :", X_layer2.shape)
print("Layer 3 :", X_layer3.shape)
print("AvgPool :", X_avgpool.shape)

# ------------------------------------------------------------
# DATASET SUMMARY
# ------------------------------------------------------------

print("\nGround truth:")
print("Knife:", np.sum(y == 1))
print("Non-knife:", np.sum(y == 0))

print("\nResNet predictions:")
print("Knife:", np.sum(resnet_pred == 1))
print("Non-knife:", np.sum(resnet_pred == 0))

# ------------------------------------------------------------
# ERROR TYPES
# ------------------------------------------------------------

false_positive = (y == 0) & (resnet_pred == 1)
false_negative = (y == 1) & (resnet_pred == 0)

print("\nResNet errors:")
print("False positives:", false_positive.sum())
print("False negatives:", false_negative.sum())
print("Total errors:",
      (false_positive | false_negative).sum())

# ------------------------------------------------------------
# SAME SPLIT AS EDGES + SILHOUETTES
# ------------------------------------------------------------

indices = np.arange(len(df))

train_idx, test_idx = train_test_split(
    indices,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("\nTrain size:", len(train_idx))
print("Test size :", len(test_idx))

print(
    "Train knives:",
    y[train_idx].sum(),
    "| Train non-knives:",
    (y[train_idx] == 0).sum()
)

print(
    "Test knives:",
    y[test_idx].sum(),
    "| Test non-knives:",
    (y[test_idx] == 0).sum()
)

# ------------------------------------------------------------
# PROBE FUNCTION
# ------------------------------------------------------------

def run_probe(X, name):

    clf = LogisticRegression(
        class_weight="balanced",
        max_iter=5000,
        random_state=42
    )

    clf.fit(
        X[train_idx],
        y[train_idx]
    )

    probe_pred = clf.predict(X[test_idx])

    # --------------------------------------------------------
    # OVERALL PERFORMANCE
    # --------------------------------------------------------

    overall_acc = accuracy_score(
        y[test_idx],
        probe_pred
    )

    balanced_acc = balanced_accuracy_score(
        y[test_idx],
        probe_pred
    )

    cm = confusion_matrix(
        y[test_idx],
        probe_pred,
        labels=[0, 1]
    )

    # --------------------------------------------------------
    # RESNET ERROR SUBSET
    # --------------------------------------------------------

    test_resnet_error = (
        false_positive[test_idx] |
        false_negative[test_idx]
    )

    error_idx = test_idx[test_resnet_error]

    if len(error_idx) > 0:

        error_pred = clf.predict(X[error_idx])

        error_acc = accuracy_score(
            y[error_idx],
            error_pred
        )

        error_balanced = balanced_accuracy_score(
            y[error_idx],
            error_pred
        )

        error_cm = confusion_matrix(
            y[error_idx],
            error_pred,
            labels=[0, 1]
        )

    else:

        error_acc = np.nan
        error_balanced = np.nan
        error_cm = np.array([])

    # --------------------------------------------------------
    # FALSE POSITIVE SUBSET
    # --------------------------------------------------------

    fp_idx = test_idx[false_positive[test_idx]]

    if len(fp_idx) > 0:

        fp_pred = clf.predict(X[fp_idx])

        fp_acc = accuracy_score(
            y[fp_idx],
            fp_pred
        )

    else:
        fp_acc = np.nan

    # --------------------------------------------------------
    # FALSE NEGATIVE SUBSET
    # --------------------------------------------------------

    fn_idx = test_idx[false_negative[test_idx]]

    if len(fn_idx) > 0:

        fn_pred = clf.predict(X[fn_idx])

        fn_acc = accuracy_score(
            y[fn_idx],
            fn_pred
        )

    else:
        fn_acc = np.nan

    # --------------------------------------------------------
    # PRINT
    # --------------------------------------------------------

    print("\n============================================")
    print(name)
    print("============================================")

    print(
        "Overall accuracy:",
        round(overall_acc, 4)
    )

    print(
        "Balanced accuracy:",
        round(balanced_acc, 4)
    )

    print("\nOverall confusion matrix:")
    print(cm)

    print("\nResNet errors in test set:", len(error_idx))

    if len(error_idx) > 0:

        print(
            "Probe accuracy on ResNet errors:",
            round(error_acc, 4)
        )

        print(
            "Probe balanced accuracy on errors:",
            round(error_balanced, 4)
        )

        print("\nError confusion matrix:")
        print(error_cm)

    print("\nFalse positives in test:", len(fp_idx))

    if len(fp_idx) > 0:
        print(
            "Probe accuracy on false positives:",
            round(fp_acc, 4)
        )

    print("\nFalse negatives in test:", len(fn_idx))

    if len(fn_idx) > 0:
        print(
            "Probe accuracy on false negatives:",
            round(fn_acc, 4)
        )

    return {
        "representation": name,
        "overall_accuracy": overall_acc,
        "balanced_accuracy": balanced_acc,
        "resnet_errors_test": len(error_idx),
        "error_accuracy": error_acc,
        "error_balanced_accuracy": error_balanced,
        "false_positive_test": len(fp_idx),
        "false_positive_accuracy": fp_acc,
        "false_negative_test": len(fn_idx),
        "false_negative_accuracy": fn_acc
    }

# ------------------------------------------------------------
# RUN ALL THREE REPRESENTATIONS
# ------------------------------------------------------------

results = []

results.append(
    run_probe(
        X_layer2,
        "CUE-CONFLICT — LAYER 2"
    )
)

results.append(
    run_probe(
        X_layer3,
        "CUE-CONFLICT — LAYER 3"
    )
)

results.append(
    run_probe(
        X_avgpool,
        "CUE-CONFLICT — AVGPOOL"
    )

)

# ------------------------------------------------------------
# SUMMARY TABLE
# ------------------------------------------------------------

results_df = pd.DataFrame(results)

print("\n\n============================================================")
print("CUE-CONFLICT DIRECT LINEAR PROBE — FINAL SUMMARY")
print("============================================================")

print(
    results_df.to_string(index=False)
)
```

```python
# ============================================================
# REPEATED STRATIFIED CROSS-VALIDATION
# DIRECT LINEAR PROBE
#
# 3 CONDITIONS × 3 REPRESENTATIONS
# 5 folds × 5 repeats
# ============================================================

import os
import json
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score
)

# ============================================================
# SETTINGS
# ============================================================

N_SPLITS = 5
N_REPEATS = 5

seeds = [42, 123, 2024, 7, 99]

# ============================================================
# HELPER FUNCTION
# ============================================================

def repeated_probe(
    X,
    y,
    resnet_pred,
    representation_name,
    condition_name
):

    records = []

    false_positive = (
        (y == 0) &
        (resnet_pred == 1)
    )

    false_negative = (
        (y == 1) &
        (resnet_pred == 0)
    )

    for repeat, seed in enumerate(seeds, start=1):

        cv = StratifiedKFold(
            n_splits=N_SPLITS,
            shuffle=True,
            random_state=seed
        )

        for fold, (train_idx, test_idx) in enumerate(
            cv.split(X, y),
            start=1
        ):

            # ------------------------------------------------
            # TRAIN LINEAR PROBE
            # ------------------------------------------------

            clf = LogisticRegression(
                class_weight="balanced",
                max_iter=5000,
                random_state=42
            )

            clf.fit(
                X[train_idx],
                y[train_idx]
            )

            # ------------------------------------------------
            # PREDICT HELD-OUT DATA
            # ------------------------------------------------

            probe_pred = clf.predict(
                X[test_idx]
            )

            # ------------------------------------------------
            # OVERALL
            # ------------------------------------------------

            overall_acc = accuracy_score(
                y[test_idx],
                probe_pred
            )

            balanced_acc = balanced_accuracy_score(
                y[test_idx],
                probe_pred
            )

            # ------------------------------------------------
            # RESNET ERROR SUBSETS
            # ------------------------------------------------

            fp_test = (
                false_positive[test_idx]
            )

            fn_test = (
                false_negative[test_idx]
            )

            # ------------------------------------------------
            # FALSE POSITIVE ACCURACY
            # ------------------------------------------------

            fp_idx = test_idx[fp_test]

            if len(fp_idx) > 0:

                fp_acc = accuracy_score(
                    y[fp_idx],
                    clf.predict(X[fp_idx])
                )

            else:

                fp_acc = np.nan

            # ------------------------------------------------
            # FALSE NEGATIVE ACCURACY
            # ------------------------------------------------

            fn_idx = test_idx[fn_test]

            if len(fn_idx) > 0:

                fn_acc = accuracy_score(
                    y[fn_idx],
                    clf.predict(X[fn_idx])
                )

            else:

                fn_acc = np.nan

            # ------------------------------------------------
            # ALL RESNET ERRORS
            # ------------------------------------------------

            error_mask = (
                fp_test |
                fn_test
            )

            error_idx = test_idx[error_mask]

            if len(error_idx) > 0:

                error_acc = accuracy_score(
                    y[error_idx],
                    clf.predict(X[error_idx])
                )

            else:

                error_acc = np.nan

            # ------------------------------------------------
            # STORE
            # ------------------------------------------------

            records.append({
                "condition": condition_name,
                "representation": representation_name,
                "repeat": repeat,
                "fold": fold,
                "overall_accuracy": overall_acc,
                "balanced_accuracy": balanced_acc,
                "resnet_error_accuracy": error_acc,
                "false_positive_accuracy": fp_acc,
                "false_negative_accuracy": fn_acc,
                "n_test": len(test_idx),
                "n_resnet_errors": len(error_idx),
                "n_false_positives": len(fp_idx),
                "n_false_negatives": len(fn_idx)
            })

    return pd.DataFrame(records)


# ============================================================
# LOAD DATASETS + EMBEDDINGS
# ============================================================

datasets = {}

# ------------------------------------------------------------
# EDGES
# ------------------------------------------------------------

edges_csv = (
    "/content/drive/MyDrive/"
    "original_edges_trial_data_FIXED.csv"
)

edges_embed_dir = (
    "/content/drive/MyDrive/"
    "AI_Mistakes_Error_Framework"
)

df_edges = pd.read_csv(edges_csv)

df_edges["true_label"] = (
    df_edges["category"]
    .astype(str)
    .str.lower()
    .eq("knife")
    .astype(int)
)

df_edges["resnet_pred"] = (
    df_edges["object_response"]
    .astype(str)
    .str.lower()
    .eq("knife")
    .astype(int)
)

df_edges["embedding_key"] = (
    df_edges["imagename"]
    .apply(
        lambda x:
        os.path.splitext(x)[0] + "_edges.png"
    )
)

with open(
    os.path.join(
        edges_embed_dir,
        "layer2_edges_embeddings.json"
    ),
    "r"
) as f:
    edges_l2 = json.load(f)

with open(
    os.path.join(
        edges_embed_dir,
        "layer3_edges_embeddings.json"
    ),
    "r"
) as f:
    edges_l3 = json.load(f)

with open(
    os.path.join(
        edges_embed_dir,
        "edges_embeddings.json"
    ),
    "r"
) as f:
    edges_avg = json.load(f)

datasets["Edges"] = {
    "df": df_edges,
    "Layer 2": np.array([
        edges_l2[k]
        for k in df_edges["embedding_key"]
    ]),
    "Layer 3": np.array([
        edges_l3[k]
        for k in df_edges["embedding_key"]
    ]),
    "AvgPool": np.array([
        edges_avg[k]
        for k in df_edges["embedding_key"]
    ])
}


# ------------------------------------------------------------
# SILHOUETTES
# ------------------------------------------------------------

sil_csv = (
    "/content/drive/MyDrive/"
    "black_filled_silhouettes_trial_data.csv"
)

sil_embed_dir = (
    "/content/drive/MyDrive/"
    "AI_Mistakes_Error_Framework"
)

df_sil = pd.read_csv(sil_csv)

df_sil["true_label"] = (
    df_sil["category"]
    .astype(str)
    .str.lower()
    .eq("knife")
    .astype(int)
)

df_sil["resnet_pred"] = (
    df_sil["object_response"]
    .astype(str)
    .str.lower()
    .eq("knife")
    .astype(int)
)

df_sil["embedding_key"] = (
    df_sil["imagename"]
    .astype(str)
)

with open(
    os.path.join(
        sil_embed_dir,
        "layer2_silhouette_embeddings.json"
    ),
    "r"
) as f:
    sil_l2 = json.load(f)

with open(
    os.path.join(
        sil_embed_dir,
        "layer3_silhouette_embeddings.json"
    ),
    "r"
) as f:
    sil_l3 = json.load(f)

with open(
    os.path.join(
        sil_embed_dir,
        "silhouette_embeddings.json"
    ),
    "r"
) as f:
    sil_avg = json.load(f)

datasets["Silhouettes"] = {
    "df": df_sil,
    "Layer 2": np.array([
        sil_l2[k]
        for k in df_sil["embedding_key"]
    ]),
    "Layer 3": np.array([
        sil_l3[k]
        for k in df_sil["embedding_key"]
    ]),
    "AvgPool": np.array([
        sil_avg[k]
        for k in df_sil["embedding_key"]
    ])
}


# ------------------------------------------------------------
# CUE-CONFLICT
# ------------------------------------------------------------

cue_csv = (
    "/content/drive/MyDrive/"
    "AI_Mistakes_Error_Framework/trial_data/"
    "original_cue-conflict_trial_data.csv"
)

cue_embed_dir = (
    "/content/drive/MyDrive/"
    "false_knife_analysis/"
    "cue_conflict_embeddings"
)

df_cue = pd.read_csv(cue_csv)

df_cue["true_label"] = (
    df_cue["category"]
    .astype(str)
    .str.lower()
    .eq("knife")
    .astype(int)
)

df_cue["resnet_pred"] = (
    df_cue["object_response"]
    .astype(str)
    .str.lower()
    .eq("knife")
    .astype(int)
)

df_cue["embedding_key"] = (
    df_cue["imagename"]
    .astype(str)
)

with open(
    os.path.join(
        cue_embed_dir,
        "layer2_cue_conflict_embeddings.json"
    ),
    "r"
) as f:
    cue_l2 = json.load(f)

with open(
    os.path.join(
        cue_embed_dir,
        "layer3_cue_conflict_embeddings.json"
    ),
    "r"
) as f:
    cue_l3 = json.load(f)

with open(
    os.path.join(
        cue_embed_dir,
        "avgpool_cue_conflict_embeddings.json"
    ),
    "r"
) as f:
    cue_avg = json.load(f)

datasets["Cue-conflict"] = {
    "df": df_cue,
    "Layer 2": np.array([
        cue_l2[k]
        for k in df_cue["embedding_key"]
    ]),
    "Layer 3": np.array([
        cue_l3[k]
        for k in df_cue["embedding_key"]
    ]),
    "AvgPool": np.array([
        cue_avg[k]
        for k in df_cue["embedding_key"]
    ])
}


# ============================================================
# VERIFY DATA BEFORE RUNNING
# ============================================================

print("============================================================")
print("DATA VERIFICATION")
print("============================================================")

for condition, data in datasets.items():

    df = data["df"]

    print(
        f"\n{condition}: "
        f"{len(df)} images | "
        f"Knives = {df['true_label'].sum()} | "
        f"Non-knives = {(df['true_label'] == 0).sum()}"
    )

    for rep in ["Layer 2", "Layer 3", "AvgPool"]:
        print(
            f"  {rep:8s}: "
            f"{data[rep].shape}"
        )


# ============================================================
# RUN ALL PROBES
# ============================================================

all_results = []

for condition, data in datasets.items():

    df = data["df"]

    y = df["true_label"].values
    resnet_pred = df["resnet_pred"].values

    print("\n")
    print("============================================================")
    print("RUNNING:", condition)
    print("============================================================")

    for representation in [
        "Layer 2",
        "Layer 3",
        "AvgPool"
    ]:

        print(
            f"  {representation}..."
        )

        result = repeated_probe(
            X=data[representation],
            y=y,
            resnet_pred=resnet_pred,
            representation_name=representation,
            condition_name=condition
        )

        all_results.append(result)


# ============================================================
# COMBINE
# ============================================================

results_df = pd.concat(
    all_results,
    ignore_index=True
)


# ============================================================
# SUMMARY: MEAN ± SD
# ============================================================

summary_rows = []

for (condition, representation), group in results_df.groupby(
    ["condition", "representation"]
):

    summary_rows.append({

        "condition": condition,

        "representation": representation,

        "overall_accuracy_mean":
            group["overall_accuracy"].mean(),

        "overall_accuracy_sd":
            group["overall_accuracy"].std(),

        "balanced_accuracy_mean":
            group["balanced_accuracy"].mean(),

        "balanced_accuracy_sd":
            group["balanced_accuracy"].std(),

        "resnet_error_accuracy_mean":
            group["resnet_error_accuracy"].mean(),

        "resnet_error_accuracy_sd":
            group["resnet_error_accuracy"].std(),

        "false_positive_accuracy_mean":
            group["false_positive_accuracy"].mean(),

        "false_positive_accuracy_sd":
            group["false_positive_accuracy"].std(),

        "false_negative_accuracy_mean":
            group["false_negative_accuracy"].mean(),

        "false_negative_accuracy_sd":
            group["false_negative_accuracy"].std()
    })

summary_df = pd.DataFrame(summary_rows)


# ============================================================
# PRINT FINAL SUMMARY
# ============================================================

print("\n\n")
print("============================================================")
print("REPEATED CROSS-VALIDATION — FINAL SUMMARY")
print("============================================================")

for _, row in summary_df.iterrows():

    print("\n--------------------------------------------")

    print(
        row["condition"],
        "—",
        row["representation"]
    )

    print(
        "Overall accuracy:",
        f"{row['overall_accuracy_mean']:.4f}",
        "±",
        f"{row['overall_accuracy_sd']:.4f}"
    )

    print(
        "Balanced accuracy:",
        f"{row['balanced_accuracy_mean']:.4f}",
        "±",
        f"{row['balanced_accuracy_sd']:.4f}"
    )

    print(
        "ResNet-error accuracy:",
        f"{row['resnet_error_accuracy_mean']:.4f}",
        "±",
        f"{row['resnet_error_accuracy_sd']:.4f}"
    )

    print(
        "False-positive accuracy:",
        f"{row['false_positive_accuracy_mean']:.4f}",
        "±",
        f"{row['false_positive_accuracy_sd']:.4f}"
    )

    print(
        "False-negative accuracy:",
        f"{row['false_negative_accuracy_mean']:.4f}",
        "±",
        f"{row['false_negative_accuracy_sd']:.4f}"
    )


# ============================================================
# SAVE RESULTS
# ============================================================

save_dir = (
    "/content/drive/MyDrive/"
    "false_knife_analysis/"
    "direct_linear_probe"
)

os.makedirs(save_dir, exist_ok=True)

results_df.to_csv(
    os.path.join(
        save_dir,
        "repeated_cv_all_folds.csv"
    ),
    index=False
)

summary_df.to_csv(
    os.path.join(
        save_dir,
        "repeated_cv_summary.csv"
    ),
    index=False
)

print("\n============================================================")
print("SAVED")
print("============================================================")

print(
    os.path.join(
        save_dir,
        "repeated_cv_all_folds.csv"
    )
)

print(
    os.path.join(
        save_dir,
        "repeated_cv_summary.csv"
    )
)
```

```python
# ============================================================
# REPEATED STRATIFIED CROSS-VALIDATION
# DIRECT LINEAR PROBE
#
# 3 CONDITIONS × 3 REPRESENTATIONS
# 5 folds × 5 repeats
# ============================================================

import os
import json
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score
)

# ============================================================
# SETTINGS
# ============================================================

N_SPLITS = 5
N_REPEATS = 5

seeds = [42, 123, 2024, 7, 99]

# ============================================================
# HELPER FUNCTION
# ============================================================

def repeated_probe(
    X,
    y,
    resnet_pred,
    representation_name,
    condition_name
):

    records = []

    false_positive = (
        (y == 0) &
        (resnet_pred == 1)
    )

    false_negative = (
        (y == 1) &
        (resnet_pred == 0)
    )

    for repeat, seed in enumerate(seeds, start=1):

        cv = StratifiedKFold(
            n_splits=N_SPLITS,
            shuffle=True,
            random_state=seed
        )

        for fold, (train_idx, test_idx) in enumerate(
            cv.split(X, y),
            start=1
        ):

            # ------------------------------------------------
            # TRAIN LINEAR PROBE
            # ------------------------------------------------

            clf = LogisticRegression(
                class_weight="balanced",
                max_iter=5000,
                random_state=42
            )

            clf.fit(
                X[train_idx],
                y[train_idx]
            )

            # ------------------------------------------------
            # PREDICT HELD-OUT DATA
            # ------------------------------------------------

            probe_pred = clf.predict(
                X[test_idx]
            )

            # ------------------------------------------------
            # OVERALL
            # ------------------------------------------------

            overall_acc = accuracy_score(
                y[test_idx],
                probe_pred
            )

            balanced_acc = balanced_accuracy_score(
                y[test_idx],
                probe_pred
            )

            # ------------------------------------------------
            # RESNET ERROR SUBSETS
            # ------------------------------------------------

            fp_test = (
                false_positive[test_idx]
            )

            fn_test = (
                false_negative[test_idx]
            )

            # ------------------------------------------------
            # FALSE POSITIVE ACCURACY
            # ------------------------------------------------

            fp_idx = test_idx[fp_test]

            if len(fp_idx) > 0:

                fp_acc = accuracy_score(
                    y[fp_idx],
                    clf.predict(X[fp_idx])
                )

            else:

                fp_acc = np.nan

            # ------------------------------------------------
            # FALSE NEGATIVE ACCURACY
            # ------------------------------------------------

            fn_idx = test_idx[fn_test]

            if len(fn_idx) > 0:

                fn_acc = accuracy_score(
                    y[fn_idx],
                    clf.predict(X[fn_idx])
                )

            else:

                fn_acc = np.nan

            # ------------------------------------------------
            # ALL RESNET ERRORS
            # ------------------------------------------------

            error_mask = (
                fp_test |
                fn_test
            )

            error_idx = test_idx[error_mask]

            if len(error_idx) > 0:

                error_acc = accuracy_score(
                    y[error_idx],
                    clf.predict(X[error_idx])
                )

            else:

                error_acc = np.nan

            # ------------------------------------------------
            # STORE
            # ------------------------------------------------

            records.append({
                "condition": condition_name,
                "representation": representation_name,
                "repeat": repeat,
                "fold": fold,
                "overall_accuracy": overall_acc,
                "balanced_accuracy": balanced_acc,
                "resnet_error_accuracy": error_acc,
                "false_positive_accuracy": fp_acc,
                "false_negative_accuracy": fn_acc,
                "n_test": len(test_idx),
                "n_resnet_errors": len(error_idx),
                "n_false_positives": len(fp_idx),
                "n_false_negatives": len(fn_idx)
            })

    return pd.DataFrame(records)


# ============================================================
# LOAD DATASETS + EMBEDDINGS
# ============================================================

datasets = {}

# ------------------------------------------------------------
# EDGES
# ------------------------------------------------------------

edges_csv = (
    "/content/drive/MyDrive/"
    "original_edges_trial_data_FIXED.csv"
)

edges_embed_dir = (
    "/content/drive/MyDrive/"
    "AI_Mistakes_Error_Framework"
)

df_edges = pd.read_csv(edges_csv)

df_edges["true_label"] = (
    df_edges["category"]
    .astype(str)
    .str.lower()
    .eq("knife")
    .astype(int)
)

df_edges["resnet_pred"] = (
    df_edges["object_response"]
    .astype(str)
    .str.lower()
    .eq("knife")
    .astype(int)
)

df_edges["embedding_key"] = (
    df_edges["imagename"]
    .apply(
        lambda x:
        os.path.splitext(x)[0] + "_edges.png"
    )
)

with open(
    os.path.join(
        edges_embed_dir,
        "layer2_edges_embeddings.json"
    ),
    "r"
) as f:
    edges_l2 = json.load(f)

with open(
    os.path.join(
        edges_embed_dir,
        "layer3_edges_embeddings.json"
    ),
    "r"
) as f:
    edges_l3 = json.load(f)

with open(
    os.path.join(
        edges_embed_dir,
        "edges_embeddings.json"
    ),
    "r"
) as f:
    edges_avg = json.load(f)

datasets["Edges"] = {
    "df": df_edges,
    "Layer 2": np.array([
        edges_l2[k]
        for k in df_edges["embedding_key"]
    ]),
    "Layer 3": np.array([
        edges_l3[k]
        for k in df_edges["embedding_key"]
    ]),
    "AvgPool": np.array([
        edges_avg[k]
        for k in df_edges["embedding_key"]
    ])
}


# ------------------------------------------------------------
# SILHOUETTES
# ------------------------------------------------------------

sil_csv = (
    "/content/drive/MyDrive/"
    "black_filled_silhouettes_trial_data.csv"
)

sil_embed_dir = (
    "/content/drive/MyDrive/"
    "AI_Mistakes_Error_Framework"
)

df_sil = pd.read_csv(sil_csv)

df_sil["true_label"] = (
    df_sil["category"]
    .astype(str)
    .str.lower()
    .eq("knife")
    .astype(int)
)

df_sil["resnet_pred"] = (
    df_sil["object_response"]
    .astype(str)
    .str.lower()
    .eq("knife")
    .astype(int)
)

df_sil["embedding_key"] = (
    df_sil["imagename"]
    .astype(str)
)

with open(
    os.path.join(
        sil_embed_dir,
        "layer2_silhouette_embeddings.json"
    ),
    "r"
) as f:
    sil_l2 = json.load(f)

with open(
    os.path.join(
        sil_embed_dir,
        "layer3_silhouette_embeddings.json"
    ),
    "r"
) as f:
    sil_l3 = json.load(f)

with open(
    os.path.join(
        sil_embed_dir,
        "silhouette_embeddings.json"
    ),
    "r"
) as f:
    sil_avg = json.load(f)

datasets["Silhouettes"] = {
    "df": df_sil,
    "Layer 2": np.array([
        sil_l2[k]
        for k in df_sil["embedding_key"]
    ]),
    "Layer 3": np.array([
        sil_l3[k]
        for k in df_sil["embedding_key"]
    ]),
    "AvgPool": np.array([
        sil_avg[k]
        for k in df_sil["embedding_key"]
    ])
}


# ------------------------------------------------------------
# CUE-CONFLICT
# ------------------------------------------------------------

cue_csv = (
    "/content/drive/MyDrive/"
    "AI_Mistakes_Error_Framework/trial_data/"
    "original_cue-conflict_trial_data.csv"
)

cue_embed_dir = (
    "/content/drive/MyDrive/"
    "false_knife_analysis/"
    "cue_conflict_embeddings"
)

df_cue = pd.read_csv(cue_csv)

df_cue["true_label"] = (
    df_cue["category"]
    .astype(str)
    .str.lower()
    .eq("knife")
    .astype(int)
)

df_cue["resnet_pred"] = (
    df_cue["object_response"]
    .astype(str)
    .str.lower()
    .eq("knife")
    .astype(int)
)

df_cue["embedding_key"] = (
    df_cue["imagename"]
    .astype(str)
)

with open(
    os.path.join(
        cue_embed_dir,
        "layer2_cue_conflict_embeddings.json"
    ),
    "r"
) as f:
    cue_l2 = json.load(f)

with open(
    os.path.join(
        cue_embed_dir,
        "layer3_cue_conflict_embeddings.json"
    ),
    "r"
) as f:
    cue_l3 = json.load(f)

with open(
    os.path.join(
        cue_embed_dir,
        "avgpool_cue_conflict_embeddings.json"
    ),
    "r"
) as f:
    cue_avg = json.load(f)

datasets["Cue-conflict"] = {
    "df": df_cue,
    "Layer 2": np.array([
        cue_l2[k]
        for k in df_cue["embedding_key"]
    ]),
    "Layer 3": np.array([
        cue_l3[k]
        for k in df_cue["embedding_key"]
    ]),
    "AvgPool": np.array([
        cue_avg[k]
        for k in df_cue["embedding_key"]
    ])
}


# ============================================================
# VERIFY DATA BEFORE RUNNING
# ============================================================

print("============================================================")
print("DATA VERIFICATION")
print("============================================================")

for condition, data in datasets.items():

    df = data["df"]

    print(
        f"\n{condition}: "
        f"{len(df)} images | "
        f"Knives = {df['true_label'].sum()} | "
        f"Non-knives = {(df['true_label'] == 0).sum()}"
    )

    for rep in ["Layer 2", "Layer 3", "AvgPool"]:
        print(
            f"  {rep:8s}: "
            f"{data[rep].shape}"
        )


# ============================================================
# RUN ALL PROBES
# ============================================================

all_results = []

for condition, data in datasets.items():

    df = data["df"]

    y = df["true_label"].values
    resnet_pred = df["resnet_pred"].values

    print("\n")
    print("============================================================")
    print("RUNNING:", condition)
    print("============================================================")

    for representation in [
        "Layer 2",
        "Layer 3",
        "AvgPool"
    ]:

        print(
            f"  {representation}..."
        )

        result = repeated_probe(
            X=data[representation],
            y=y,
            resnet_pred=resnet_pred,
            representation_name=representation,
            condition_name=condition
        )

        all_results.append(result)


# ============================================================
# COMBINE
# ============================================================

results_df = pd.concat(
    all_results,
    ignore_index=True
)


# ============================================================
# SUMMARY: MEAN ± SD
# ============================================================

summary_rows = []

for (condition, representation), group in results_df.groupby(
    ["condition", "representation"]
):

    summary_rows.append({

        "condition": condition,

        "representation": representation,

        "overall_accuracy_mean":
            group["overall_accuracy"].mean(),

        "overall_accuracy_sd":
            group["overall_accuracy"].std(),

        "balanced_accuracy_mean":
            group["balanced_accuracy"].mean(),

        "balanced_accuracy_sd":
            group["balanced_accuracy"].std(),

        "resnet_error_accuracy_mean":
            group["resnet_error_accuracy"].mean(),

        "resnet_error_accuracy_sd":
            group["resnet_error_accuracy"].std(),

        "false_positive_accuracy_mean":
            group["false_positive_accuracy"].mean(),

        "false_positive_accuracy_sd":
            group["false_positive_accuracy"].std(),

        "false_negative_accuracy_mean":
            group["false_negative_accuracy"].mean(),

        "false_negative_accuracy_sd":
            group["false_negative_accuracy"].std()
    })

summary_df = pd.DataFrame(summary_rows)


# ============================================================
# PRINT FINAL SUMMARY
# ============================================================

print("\n\n")
print("============================================================")
print("REPEATED CROSS-VALIDATION — FINAL SUMMARY")
print("============================================================")

for _, row in summary_df.iterrows():

    print("\n--------------------------------------------")

    print(
        row["condition"],
        "—",
        row["representation"]
    )

    print(
        "Overall accuracy:",
        f"{row['overall_accuracy_mean']:.4f}",
        "±",
        f"{row['overall_accuracy_sd']:.4f}"
    )

    print(
        "Balanced accuracy:",
        f"{row['balanced_accuracy_mean']:.4f}",
        "±",
        f"{row['balanced_accuracy_sd']:.4f}"
    )

    print(
        "ResNet-error accuracy:",
        f"{row['resnet_error_accuracy_mean']:.4f}",
        "±",
        f"{row['resnet_error_accuracy_sd']:.4f}"
    )

    print(
        "False-positive accuracy:",
        f"{row['false_positive_accuracy_mean']:.4f}",
        "±",
        f"{row['false_positive_accuracy_sd']:.4f}"
    )

    print(
        "False-negative accuracy:",
        f"{row['false_negative_accuracy_mean']:.4f}",
        "±",
        f"{row['false_negative_accuracy_sd']:.4f}"
    )


# ============================================================
# SAVE RESULTS
# ============================================================

save_dir = (
    "/content/drive/MyDrive/"
    "false_knife_analysis/"
    "direct_linear_probe"
)

os.makedirs(save_dir, exist_ok=True)

results_df.to_csv(
    os.path.join(
        save_dir,
        "repeated_cv_all_folds.csv"
    ),
    index=False
)

summary_df.to_csv(
    os.path.join(
        save_dir,
        "repeated_cv_summary.csv"
    ),
    index=False
)

print("\n============================================================")
print("SAVED")
print("============================================================")

print(
    os.path.join(
        save_dir,
        "repeated_cv_all_folds.csv"
    )
)

print(
    os.path.join(
        save_dir,
        "repeated_cv_summary.csv"
    )
)
```

```python
# ============================================================
# DIRECT LINEAR PROBE — REPEATED STRATIFIED CROSS-VALIDATION
# 5 FOLDS × 5 REPEATS = 25 HELD-OUT EVALUATIONS
#
# Conditions:
#   1. Edges
#   2. Silhouettes
#   3. Cue-conflict
#
# Representations:
#   1. Layer 2
#   2. Layer 3
#   3. AvgPool
#
# IMPORTANT:
# Edges AvgPool keys contain:
#     category/filename
# e.g. airplane/airplane1_edges.png
# while CSV uses:
#     airplane1_edges.png
# ============================================================

import os
import json
import numpy as np
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, balanced_accuracy_score


# ============================================================
# 1. PATHS
# ============================================================

BASE = "/content/drive/MyDrive/false_knife_analysis"

AI_BASE = "/content/drive/MyDrive/AI_Mistakes_Error_Framework"

# CSVs
EDGES_CSV = "/content/drive/MyDrive/original_edges_trial_data_FIXED.csv"

SILHOUETTES_CSV = "/content/drive/MyDrive/black_filled_silhouettes_trial_data.csv"

CUE_CSV = (
    "/content/drive/MyDrive/"
    "AI_Mistakes_Error_Framework/trial_data/"
    "original_cue-conflict_trial_data.csv"
)

# -------------------------
# Edges embeddings
# -------------------------

EDGES_L2 = (
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/"
    "layer2_edges_embeddings.json"
)

EDGES_L3 = (
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/"
    "layer3_edges_embeddings.json"
)

EDGES_AVG = (
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/"
    "edges_embeddings.json"
)

# -------------------------
# Silhouettes embeddings
# -------------------------

SIL_L2 = (
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/"
    "layer2_silhouette_embeddings.json"
)

SIL_L3 = (
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/"
    "layer3_silhouette_embeddings.json"
)

SIL_AVG = (
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework/"
    "silhouette_embeddings.json"
)

# -------------------------
# Cue-conflict embeddings
# -------------------------

CUE_L2 = (
    "/content/drive/MyDrive/false_knife_analysis/"
    "cue_conflict_embeddings/"
    "layer2_cue_conflict_embeddings.json"
)

CUE_L3 = (
    "/content/drive/MyDrive/false_knife_analysis/"
    "cue_conflict_embeddings/"
    "layer3_cue_conflict_embeddings.json"
)

CUE_AVG = (
    "/content/drive/MyDrive/false_knife_analysis/"
    "cue_conflict_embeddings/"
    "avgpool_cue_conflict_embeddings.json"
)


# ============================================================
# 2. LOAD JSON EMBEDDINGS
# ============================================================

def load_json(path):
    print("Loading:", path)
    with open(path, "r") as f:
        return json.load(f)


edges_l2 = load_json(EDGES_L2)
edges_l3 = load_json(EDGES_L3)
edges_avg = load_json(EDGES_AVG)

sil_l2 = load_json(SIL_L2)
sil_l3 = load_json(SIL_L3)
sil_avg = load_json(SIL_AVG)

cue_l2 = load_json(CUE_L2)
cue_l3 = load_json(CUE_L3)
cue_avg = load_json(CUE_AVG)


# ============================================================
# 3. LOAD CSVs
# ============================================================

df_edges = pd.read_csv(EDGES_CSV)
df_sil = pd.read_csv(SILHOUETTES_CSV)
df_cue = pd.read_csv(CUE_CSV)

print("\n============================================================")
print("DATASET SIZES")
print("============================================================")

print("Edges:", len(df_edges))
print("Silhouettes:", len(df_sil))
print("Cue-conflict:", len(df_cue))


# ============================================================
# 4. CREATE MATCHING KEYS
# ============================================================

# -------------------------
# EDGES
# -------------------------

df_edges["embedding_key"] = df_edges["imagename"].apply(
    lambda x: os.path.splitext(x)[0] + "_edges.png"
)

# -------------------------
# SILHOUETTES
# -------------------------

df_sil["embedding_key"] = df_sil["imagename"]

# -------------------------
# CUE-CONFLICT
# -------------------------

df_cue["embedding_key"] = df_cue["imagename"]


# ============================================================
# 5. CREATE TRUE LABELS
#
# knife = 1
# non-knife = 0
# ============================================================

df_edges["true_label"] = (
    df_edges["category"].str.lower() == "knife"
).astype(int)

df_sil["true_label"] = (
    df_sil["category"].str.lower() == "knife"
).astype(int)

df_cue["true_label"] = (
    df_cue["category"].str.lower() == "knife"
).astype(int)


# ============================================================
# 6. CREATE RESNET BINARY PREDICTIONS
#
# object_response == knife -> 1
# otherwise -> 0
# ============================================================

df_edges["resnet_pred"] = (
    df_edges["object_response"].astype(str).str.lower() == "knife"
).astype(int)

df_sil["resnet_pred"] = (
    df_sil["object_response"].astype(str).str.lower() == "knife"
).astype(int)

df_cue["resnet_pred"] = (
    df_cue["object_response"].astype(str).str.lower() == "knife"
).astype(int)


# ============================================================
# 7. CHECK DATA
# ============================================================

print("\n============================================================")
print("GROUND TRUTH COUNTS")
print("============================================================")

for name, df in [
    ("Edges", df_edges),
    ("Silhouettes", df_sil),
    ("Cue-conflict", df_cue)
]:
    print(
        f"{name}:",
        "knife =", int(df["true_label"].sum()),
        "| non-knife =", int((df["true_label"] == 0).sum())
    )


print("\n============================================================")
print("RESNET BINARY PREDICTION COUNTS")
print("============================================================")

for name, df in [
    ("Edges", df_edges),
    ("Silhouettes", df_sil),
    ("Cue-conflict", df_cue)
]:
    print(
        f"{name}:",
        "predicted knife =", int(df["resnet_pred"].sum()),
        "| predicted non-knife =",
        int((df["resnet_pred"] == 0).sum())
    )


# ============================================================
# 8. BUILD EMBEDDING MATRICES
# ============================================================

print("\n============================================================")
print("BUILDING EMBEDDING MATRICES")
print("============================================================")


# ------------------------------------------------------------
# EDGES
# ------------------------------------------------------------

# Layer 2
edges_X_l2 = np.array([
    edges_l2[k]
    for k in df_edges["embedding_key"]
])

# Layer 3
edges_X_l3 = np.array([
    edges_l3[k]
    for k in df_edges["embedding_key"]
])

# IMPORTANT FIX:
# AvgPool keys are:
# category/filename
#
# e.g.
# airplane/airplane1_edges.png
#
# CSV key is:
# airplane1_edges.png

edges_avg_keys = [
    f"{row['category']}/{row['embedding_key']}"
    for _, row in df_edges.iterrows()
]

# Verify all keys
missing_edges_avg = [
    k for k in edges_avg_keys
    if k not in edges_avg
]

print("\nEdges AvgPool key check:")
print("Requested:", len(edges_avg_keys))
print("Missing:", len(missing_edges_avg))

if missing_edges_avg:
    print("First missing keys:")
    for k in missing_edges_avg[:10]:
        print(k)
    raise KeyError(
        "Edges AvgPool still has missing embedding keys."
    )

edges_X_avg = np.array([
    edges_avg[k]
    for k in edges_avg_keys
])


# ------------------------------------------------------------
# SILHOUETTES
# ------------------------------------------------------------

sil_X_l2 = np.array([
    sil_l2[k]
    for k in df_sil["embedding_key"]
])

sil_X_l3 = np.array([
    sil_l3[k]
    for k in df_sil["embedding_key"]
])

sil_X_avg = np.array([
    sil_avg[k]
    for k in df_sil["embedding_key"]
])


# ------------------------------------------------------------
# CUE-CONFLICT
# ------------------------------------------------------------

cue_X_l2 = np.array([
    cue_l2[k]
    for k in df_cue["embedding_key"]
])

cue_X_l3 = np.array([
    cue_l3[k]
    for k in df_cue["embedding_key"]
])

cue_X_avg = np.array([
    cue_avg[k]
    for k in df_cue["embedding_key"]
])


# ============================================================
# 9. VERIFY MATRICES
# ============================================================

print("\n============================================================")
print("EMBEDDING MATRIX SHAPES")
print("============================================================")

print("\nEDGES")
print("Layer 2 :", edges_X_l2.shape)
print("Layer 3 :", edges_X_l3.shape)
print("AvgPool :", edges_X_avg.shape)

print("\nSILHOUETTES")
print("Layer 2 :", sil_X_l2.shape)
print("Layer 3 :", sil_X_l3.shape)
print("AvgPool :", sil_X_avg.shape)

print("\nCUE-CONFLICT")
print("Layer 2 :", cue_X_l2.shape)
print("Layer 3 :", cue_X_l3.shape)
print("AvgPool :", cue_X_avg.shape)


# ============================================================
# 10. ORGANIZE DATASETS
# ============================================================

datasets = {

    "Edges": {
        "df": df_edges,
        "Layer 2": edges_X_l2,
        "Layer 3": edges_X_l3,
        "AvgPool": edges_X_avg
    },

    "Silhouettes": {
        "df": df_sil,
        "Layer 2": sil_X_l2,
        "Layer 3": sil_X_l3,
        "AvgPool": sil_X_avg
    },

    "Cue-conflict": {
        "df": df_cue,
        "Layer 2": cue_X_l2,
        "Layer 3": cue_X_l3,
        "AvgPool": cue_X_avg
    }
}


# ============================================================
# 11. REPEATED STRATIFIED CROSS-VALIDATION
#
# 5 folds × 5 repeats
# = 25 held-out evaluations per representation
# ============================================================

N_SPLITS = 5

SEEDS = [42, 123, 2024, 7, 99]

all_results = []


for condition, data in datasets.items():

    df = data["df"]

    y = df["true_label"].to_numpy()
    resnet_pred = df["resnet_pred"].to_numpy()

    print("\n")
    print("============================================================")
    print("CONDITION:", condition.upper())
    print("============================================================")

    for representation in [
        "Layer 2",
        "Layer 3",
        "AvgPool"
    ]:

        X = data[representation]

        print(
            f"\nRunning {representation} "
            f"| X shape = {X.shape}"
        )

        for repeat_id, seed in enumerate(SEEDS, start=1):

            skf = StratifiedKFold(
                n_splits=N_SPLITS,
                shuffle=True,
                random_state=seed
            )

            for fold_id, (train_idx, test_idx) in enumerate(
                skf.split(X, y),
                start=1
            ):

                X_train = X[train_idx]
                X_test = X[test_idx]

                y_train = y[train_idx]
                y_test = y[test_idx]

                # ------------------------------------------------
                # TRAIN LINEAR PROBE
                # ------------------------------------------------

                probe = LogisticRegression(
                    class_weight="balanced",
                    max_iter=2000,
                    random_state=seed
                )

                probe.fit(
                    X_train,
                    y_train
                )

                probe_pred = probe.predict(X_test)

                # ------------------------------------------------
                # OVERALL PERFORMANCE
                # ------------------------------------------------

                overall_accuracy = accuracy_score(
                    y_test,
                    probe_pred
                )

                balanced_accuracy = balanced_accuracy_score(
                    y_test,
                    probe_pred
                )

                # ------------------------------------------------
                # RESNET ERROR SUBSET
                #
                # Only test-set cases where the original
                # ResNet binary prediction was incorrect.
                # ------------------------------------------------

                resnet_errors = (
                    resnet_pred[test_idx] != y_test
                )

                n_resnet_errors = int(
                    resnet_errors.sum()
                )

                if n_resnet_errors > 0:

                    error_true = y_test[
                        resnet_errors
                    ]

                    error_pred = probe_pred[
                        resnet_errors
                    ]

                    error_accuracy = accuracy_score(
                        error_true,
                        error_pred
                    )

                    # Balanced accuracy requires both
                    # classes to be present.
                    if len(np.unique(error_true)) == 2:

                        error_balanced_accuracy = (
                            balanced_accuracy_score(
                                error_true,
                                error_pred
                            )
                        )

                    else:
                        error_balanced_accuracy = np.nan

                else:

                    error_accuracy = np.nan
                    error_balanced_accuracy = np.nan

                # ------------------------------------------------
                # FALSE POSITIVE SUBSET
                #
                # Ground truth = non-knife
                # ResNet predicted = knife
                # ------------------------------------------------

                false_positive_mask = (
                    (y_test == 0) &
                    (resnet_pred[test_idx] == 1)
                )

                n_false_positives = int(
                    false_positive_mask.sum()
                )

                if n_false_positives > 0:

                    fp_true = y_test[
                        false_positive_mask
                    ]

                    fp_pred = probe_pred[
                        false_positive_mask
                    ]

                    false_positive_accuracy = accuracy_score(
                        fp_true,
                        fp_pred
                    )

                else:

                    false_positive_accuracy = np.nan

                # ------------------------------------------------
                # FALSE NEGATIVE SUBSET
                #
                # Ground truth = knife
                # ResNet predicted = non-knife
                # ------------------------------------------------

                false_negative_mask = (
                    (y_test == 1) &
                    (resnet_pred[test_idx] == 0)
                )

                n_false_negatives = int(
                    false_negative_mask.sum()
                )

                if n_false_negatives > 0:

                    fn_true = y_test[
                        false_negative_mask
                    ]

                    fn_pred = probe_pred[
                        false_negative_mask
                    ]

                    false_negative_accuracy = accuracy_score(
                        fn_true,
                        fn_pred
                    )

                else:

                    false_negative_accuracy = np.nan

                # ------------------------------------------------
                # SAVE FOLD RESULT
                # ------------------------------------------------

                all_results.append({

                    "condition": condition,

                    "representation": representation,

                    "repeat": repeat_id,

                    "seed": seed,

                    "fold": fold_id,

                    "n_train": len(train_idx),

                    "n_test": len(test_idx),

                    "overall_accuracy":
                        overall_accuracy,

                    "balanced_accuracy":
                        balanced_accuracy,

                    "n_resnet_errors_test":
                        n_resnet_errors,

                    "error_accuracy":
                        error_accuracy,

                    "error_balanced_accuracy":
                        error_balanced_accuracy,

                    "n_false_positives_test":
                        n_false_positives,

                    "false_positive_accuracy":
                        false_positive_accuracy,

                    "n_false_negatives_test":
                        n_false_negatives,

                    "false_negative_accuracy":
                        false_negative_accuracy
                })


# ============================================================
# 12. CONVERT TO DATAFRAME
# ============================================================

results_df = pd.DataFrame(all_results)


# ============================================================
# 13. SAVE ALL FOLD RESULTS
# ============================================================

output_dir = (
    "/content/drive/MyDrive/"
    "false_knife_analysis/direct_linear_probe"
)

os.makedirs(
    output_dir,
    exist_ok=True
)

fold_path = os.path.join(
    output_dir,
    "repeated_cv_all_folds.csv"
)

results_df.to_csv(
    fold_path,
    index=False
)


# ============================================================
# 14. CREATE FINAL SUMMARY
# ============================================================

metrics = [
    "overall_accuracy",
    "balanced_accuracy",
    "error_accuracy",
    "error_balanced_accuracy",
    "false_positive_accuracy",
    "false_negative_accuracy"
]

summary_rows = []


for (condition, representation), group in results_df.groupby(
    ["condition", "representation"]
):

    row = {
        "condition": condition,
        "representation": representation,

        "n_evaluations": len(group),

        "mean_n_resnet_errors_test":
            group["n_resnet_errors_test"].mean(),

        "sd_n_resnet_errors_test":
            group["n_resnet_errors_test"].std(),

        "mean_n_false_positives_test":
            group["n_false_positives_test"].mean(),

        "mean_n_false_negatives_test":
            group["n_false_negatives_test"].mean()
    }

    for metric in metrics:

        row[f"{metric}_mean"] = (
            group[metric].mean()
        )

        row[f"{metric}_sd"] = (
            group[metric].std()
        )

    summary_rows.append(row)


summary_df = pd.DataFrame(
    summary_rows
)


# ============================================================
# 15. SAVE SUMMARY
# ============================================================

summary_path = os.path.join(
    output_dir,
    "repeated_cv_summary.csv"
)

summary_df.to_csv(
    summary_path,
    index=False
)


# ============================================================
# 16. PRINT FINAL SUMMARY
# ============================================================

print("\n\n")
print("============================================================")
print("REPEATED CROSS-VALIDATION — FINAL SUMMARY")
print("============================================================")

for _, row in summary_df.iterrows():

    print("\n------------------------------------------------------------")

    print(
        f"{row['condition'].upper()} — "
        f"{row['representation']}"
    )

    print(
        f"Evaluations: "
        f"{int(row['n_evaluations'])}"
    )

    print(
        f"Overall accuracy: "
        f"{row['overall_accuracy_mean']:.4f} "
        f"± {row['overall_accuracy_sd']:.4f}"
    )

    print(
        f"Balanced accuracy: "
        f"{row['balanced_accuracy_mean']:.4f} "
        f"± {row['balanced_accuracy_sd']:.4f}"
    )

    print(
        f"ResNet-error accuracy: "
        f"{row['error_accuracy_mean']:.4f} "
        f"± {row['error_accuracy_sd']:.4f}"
    )

    print(
        f"ResNet-error balanced accuracy: "
        f"{row['error_balanced_accuracy_mean']:.4f} "
        f"± {row['error_balanced_accuracy_sd']:.4f}"
    )

    print(
        f"False-positive accuracy: "
        f"{row['false_positive_accuracy_mean']:.4f} "
        f"± {row['false_positive_accuracy_sd']:.4f}"
    )

    print(
        f"False-negative accuracy: "
        f"{row['false_negative_accuracy_mean']:.4f} "
        f"± {row['false_negative_accuracy_sd']:.4f}"
    )


# ============================================================
# 17. COMPACT TABLE
# ============================================================

print("\n\n")
print("============================================================")
print("COMPACT FINAL TABLE")
print("============================================================")

compact = summary_df[
    [
        "condition",
        "representation",
        "overall_accuracy_mean",
        "overall_accuracy_sd",
        "balanced_accuracy_mean",
        "balanced_accuracy_sd",
        "error_accuracy_mean",
        "error_accuracy_sd"
    ]
].copy()

compact.columns = [
    "Condition",
    "Representation",
    "Overall Mean",
    "Overall SD",
    "Balanced Mean",
    "Balanced SD",
    "ResNet Error Mean",
    "ResNet Error SD"
]

print(
    compact.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# ============================================================
# 18. FINAL FILE CHECK
# ============================================================

print("\n\n")
print("============================================================")
print("FILES SAVED")
print("============================================================")

print("All fold results:")
print(fold_path)

print("\nSummary:")
print(summary_path)

print("\nRows in fold-level results:", len(results_df))
print("Rows in summary:", len(summary_df))
```

```python

# STEP 2 — Reconstruct per-image linear-probe predictions
# Outputs: per-image probabilities, predictions, confidence, and available ResNet-error flags.

import os, json, re, glob, warnings
import numpy as np
import pandas as pd
from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.linear_model import LogisticRegression

BASE = "/content/drive/MyDrive"
OUT = f"{BASE}/false_knife_analysis/direct_linear_probe/per_image_reconstruction"
os.makedirs(OUT, exist_ok=True)

# ---------- Load files ----------
def load_json(path):
    with open(path, "r") as f:
        return json.load(f)

master_path = f"{BASE}/false_knife_analysis/false_knife_master_table.csv"
master = pd.read_csv(master_path)

cue_trials_path = f"{BASE}/original_cue-conflict_trial_data.csv"
cue_trials = pd.read_csv(cue_trials_path)

cue_pred_path = (
    f"{BASE}/false_knife_analysis/"
    "cue_conflict_alpha_intervention_47_v2/"
    "cue_conflict_alpha_intervention_47_v2_resnet_predictions.csv"
)

# Key normalizer for matching filenames across tables
def norm_name(x):
    x = os.path.basename(str(x)).lower()
    x = re.sub(r"_edges(?=\.png$)", "", x)
    return x

# ---------- Build datasets ----------
datasets = {}

# Cue-conflict: labels and original ResNet responses from trial table
cue_files = {
    "Layer2": f"{BASE}/false_knife_analysis/cue_conflict_embeddings/layer2_cue_conflict_embeddings.json",
    "Layer3": f"{BASE}/false_knife_analysis/cue_conflict_embeddings/layer3_cue_conflict_embeddings.json",
    "AvgPool": f"{BASE}/false_knife_analysis/cue_conflict_embeddings/avgpool_cue_conflict_embeddings.json",
}
cue_embs = {layer: load_json(path) for layer, path in cue_files.items()}

cue_trials["key"] = cue_trials["imagename"].map(norm_name)
cue_trials = cue_trials.drop_duplicates("key", keep="first").set_index("key")

cue_keys = sorted(set.intersection(
    *[set(d.keys()) for d in cue_embs.values()],
    set(cue_trials.index)
))
cue_meta = pd.DataFrame({
    "image_id": cue_keys,
    "category": [str(cue_trials.loc[k, "category"]).lower() for k in cue_keys],
    "resnet_response": [str(cue_trials.loc[k, "object_response"]).lower() for k in cue_keys],
})
cue_meta["y_knife"] = (cue_meta["category"] == "knife").astype(int)
cue_meta["resnet_binary_knife"] = (cue_meta["resnet_response"] == "knife").astype(int)
cue_meta["resnet_error"] = cue_meta["y_knife"] != cue_meta["resnet_binary_knife"]

datasets["Cue-conflict"] = (cue_meta, cue_embs)

# Edges and silhouettes: use labels encoded in embedding keys / label map
other_specs = {
    "Edges": {
        "Layer2": f"{BASE}/AI_Mistakes_Error_Framework/layer2_edges_embeddings.json",
        "Layer3": f"{BASE}/AI_Mistakes_Error_Framework/layer3_edges_embeddings.json",
        "AvgPool": f"{BASE}/AI_Mistakes_Error_Framework/edges_embeddings.json",
    },
    "Silhouettes": {
        "Layer2": f"{BASE}/AI_Mistakes_Error_Framework/layer2_silhouette_embeddings.json",
        "Layer3": f"{BASE}/AI_Mistakes_Error_Framework/layer3_silhouette_embeddings.json",
        "AvgPool": f"{BASE}/AI_Mistakes_Error_Framework/silhouette_embeddings.json",
    }
}

sil_labels = load_json(
    f"{BASE}/AI_Mistakes_Error_Framework/silhouette_shape_classes.json"
)

# Use the master table only where it provides known original false-knife cases.
master["match_key"] = master["imagename"].map(norm_name)
known_fp = {}
for _, r in master.iterrows():
    cond = str(r["condition"]).lower()
    if str(r["model_response"]).lower() == "knife":
        known_fp[(cond, r["match_key"])] = True

for condition, paths in other_specs.items():
    embs = {layer: load_json(path) for layer, path in paths.items()}
    common = sorted(set.intersection(*[set(d.keys()) for d in embs.values()]))

    rows = []
    for key in common:
        if condition == "Silhouettes":
            # Labels file keys are e.g. airplane1.png
            lab = sil_labels.get(key, {})
            category = str(lab.get("category", os.path.basename(key).split(".")[0])).lower()
            match_key = norm_name(key)
        else:
            # Edge embedding keys are e.g. airplane/airplane10_edges.png
            base_name = os.path.basename(key)
            category = base_name.split("_")[0].split(".")[0].lower()
            match_key = norm_name(base_name)

        # Master table records known false-knife cases, not every image.
        cond_key = "edges" if condition == "Edges" else "silhouettes"
        is_known_fp = known_fp.get((cond_key, match_key), False)

        rows.append({
            "image_id": key,
            "category": category,
            "y_knife": int(category == "knife"),
            "resnet_response": "knife" if is_known_fp else "unknown",
            "resnet_binary_knife": 1 if is_known_fp else np.nan,
            "resnet_error": bool(is_known_fp) if category != "knife" else np.nan,
            "resnet_error_status": (
                "known_false_knife" if is_known_fp
                else "unknown_from_available_master"
            ),
        })

    meta = pd.DataFrame(rows)
    if condition == "Cue-conflict":
        meta["resnet_error_status"] = "known_from_trial_table"
    datasets[condition] = (meta, embs)

# ---------- Repeated out-of-fold predictions ----------
all_results = []
summary_rows = []

for condition, (meta, embeddings) in datasets.items():
    print(f"\nProcessing {condition}: {len(meta)} images")

    for layer, emb_dict in embeddings.items():
        keys = meta["image_id"].tolist()
        X = np.asarray([emb_dict[k] for k in keys], dtype=np.float32)
        y = meta["y_knife"].to_numpy(dtype=int)

        counts = np.bincount(y, minlength=2)
        if counts.min() < 5:
            print(f"Skipping {condition}/{layer}: too few examples in a class")
            continue

        # Five folds × five repeats; each image is held out once per repeat.
        cv = RepeatedStratifiedKFold(
            n_splits=5, n_repeats=5, random_state=42
        )
        p_sum = np.zeros(len(y), dtype=float)
        n_seen = np.zeros(len(y), dtype=int)

        for train_idx, test_idx in cv.split(X, y):
            clf = LogisticRegression(
                class_weight="balanced",
                max_iter=5000,
                solver="lbfgs"
            )
            clf.fit(X[train_idx], y[train_idx])
            knife_col = list(clf.classes_).index(1)
            p = clf.predict_proba(X[test_idx])[:, knife_col]
            p_sum[test_idx] += p
            n_seen[test_idx] += 1

        p_knife = p_sum / n_seen
        yhat = (p_knife >= 0.5).astype(int)

        result = meta.copy()
        result["condition"] = condition
        result["representation"] = layer
        result["probe_p_knife_mean_oof"] = p_knife
        result["probe_p_nonknife_mean_oof"] = 1 - p_knife
        result["probe_prediction"] = np.where(yhat == 1, "knife", "non-knife")
        result["probe_correct"] = (yhat == y).astype(int)
        result["probe_confidence"] = np.maximum(p_knife, 1 - p_knife)
        result["probe_margin"] = np.abs(2 * p_knife - 1)
        result["n_oof_repeats"] = n_seen

        # ResNet-error accuracy only where original binary predictions are known.
        known = result["resnet_error"].notna()
        err = result["resnet_error"].fillna(False).astype(bool)
        error_subset = result[known & err]

        summary_rows.append({
            "condition": condition,
            "representation": layer,
            "n_images": len(result),
            "n_true_knife": int((y == 1).sum()),
            "n_true_nonknife": int((y == 0).sum()),
            "reconstructed_oof_accuracy": float((yhat == y).mean()),
            "reconstructed_oof_balanced_accuracy": float(
                (np.mean(yhat[y == 1] == 1) +
                 np.mean(yhat[y == 0] == 0)) / 2
            ),
            "known_resnet_error_images": int(len(error_subset)),
            "probe_accuracy_on_known_resnet_errors": (
                float(error_subset["probe_correct"].mean())
                if len(error_subset) else np.nan
            ),
        })

        all_results.append(result)

results = pd.concat(all_results, ignore_index=True)
summary = pd.DataFrame(summary_rows)

image_out = f"{OUT}/RECONSTRUCTED_per_image_probe_predictions.csv"
summary_out = f"{OUT}/RECONSTRUCTED_probe_summary.csv"

results.to_csv(image_out, index=False)
summary.to_csv(summary_out, index=False)

print("\n" + "=" * 75)
print("RECONSTRUCTION COMPLETE")
print("=" * 75)
print("\nSummary:")
print(summary.round(4).to_string(index=False))
print("\nSaved per-image predictions:", image_out)
print("Saved summary:", summary_out)

print("\nCue-conflict: examples where ResNet was wrong but probe favours non-knife")
cue = results[results["condition"] == "Cue-conflict"].copy()
examples = cue[
    (cue["resnet_error"] == True) &
    (cue["probe_p_nonknife_mean_oof"] > 0.5)
].sort_values("probe_p_nonknife_mean_oof", ascending=False)

print(examples[
    ["image_id", "category", "resnet_response",
     "probe_p_knife_mean_oof", "probe_p_nonknife_mean_oof",
     "probe_prediction", "probe_correct"]
].head(15).to_string(index=False))
```

```python

# FIX — Reconstruct Edges and Silhouettes only
import os, json
import numpy as np
import pandas as pd
from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.linear_model import LogisticRegression

BASE = "/content/drive/MyDrive"
OUT = f"{BASE}/false_knife_analysis/direct_linear_probe/per_image_reconstruction"
os.makedirs(OUT, exist_ok=True)

def loadj(path):
    with open(path) as f:
        return json.load(f)

def run_probe(condition, embedding_paths, label_map=None):
    embs = {layer: loadj(path) for layer, path in embedding_paths.items()}

    # Use the keys from the first embedding and keep only keys present in all layers.
    keys = sorted(set.intersection(*[set(d.keys()) for d in embs.values()]))
    print(f"\n{condition}: matched embedding keys = {len(keys)}")
    if not keys:
        print("No matched keys; check the embedding paths.")
        return None

    rows = []
    for key in keys:
        filename = os.path.basename(key)

        if condition == "Silhouettes":
            label_info = (label_map or {}).get(key, {})
            category = str(label_info.get("category", filename.split(".")[0])).lower()
        else:
            # Example edge key: airplane/airplane10_edges.png
            # Category is the folder name, not the whole filename.
            category = key.replace("\\", "/").split("/")[0].lower()

        rows.append({
            "image_id": key,
            "category": category,
            "y_knife": int(category == "knife")
        })

    meta = pd.DataFrame(rows)
    outputs = []

    for layer, emb_dict in embs.items():
        X = np.asarray([emb_dict[k] for k in meta["image_id"]], dtype=np.float32)
        y = meta["y_knife"].to_numpy(dtype=int)

        counts = np.bincount(y, minlength=2)
        print(f"{layer}: X={X.shape}; knife={counts[1]}, non-knife={counts[0]}")

        if counts.min() < 5:
            print(f"Skipping {condition}/{layer}: fewer than 5 examples in a class.")
            continue

        cv = RepeatedStratifiedKFold(n_splits=5, n_repeats=5, random_state=42)
        p_sum = np.zeros(len(y))
        n_seen = np.zeros(len(y), dtype=int)

        for train_idx, test_idx in cv.split(X, y):
            model = LogisticRegression(
                class_weight="balanced", max_iter=5000, solver="lbfgs"
            )
            model.fit(X[train_idx], y[train_idx])
            knife_col = list(model.classes_).index(1)
            p_sum[test_idx] += model.predict_proba(X[test_idx])[:, knife_col]
            n_seen[test_idx] += 1

        p_knife = p_sum / n_seen
        pred = (p_knife >= 0.5).astype(int)

        res = meta.copy()
        res["condition"] = condition
        res["representation"] = layer
        res["probe_p_knife_mean_oof"] = p_knife
        res["probe_p_nonknife_mean_oof"] = 1 - p_knife
        res["probe_prediction"] = np.where(pred == 1, "knife", "non-knife")
        res["probe_correct"] = (pred == y).astype(int)
        res["probe_confidence"] = np.maximum(p_knife, 1 - p_knife)
        res["probe_margin"] = np.abs(2 * p_knife - 1)
        res["n_oof_repeats"] = n_seen
        outputs.append(res)

    return outputs

sil_labels = loadj(
    f"{BASE}/AI_Mistakes_Error_Framework/silhouette_shape_classes.json"
)

edge_outputs = run_probe("Edges", {
    "Layer2": f"{BASE}/AI_Mistakes_Error_Framework/layer2_edges_embeddings.json",
    "Layer3": f"{BASE}/AI_Mistakes_Error_Framework/layer3_edges_embeddings.json",
    "AvgPool": f"{BASE}/AI_Mistakes_Error_Framework/edges_embeddings.json",
})

sil_outputs = run_probe("Silhouettes", {
    "Layer2": f"{BASE}/AI_Mistakes_Error_Framework/layer2_silhouette_embeddings.json",
    "Layer3": f"{BASE}/AI_Mistakes_Error_Framework/layer3_silhouette_embeddings.json",
    "AvgPool": f"{BASE}/AI_Mistakes_Error_Framework/silhouette_embeddings.json",
}, label_map=sil_labels)

for condition, outputs in [("Edges", edge_outputs), ("Silhouettes", sil_outputs)]:
    if outputs:
        combined = pd.concat(outputs, ignore_index=True)
        path = f"{OUT}/RECONSTRUCTED_{condition.lower()}_per_image.csv"
        combined.to_csv(path, index=False)
        print(f"\nSaved {condition} per-image results to:\n{path}")
        print(combined.groupby("representation")["probe_correct"].mean().round(4))
```

```python

import os, re, json
import numpy as np
import pandas as pd
from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.linear_model import LogisticRegression

OUT = "/content/drive/MyDrive/false_knife_analysis/direct_linear_probe/per_image_reconstruction"
os.makedirs(OUT, exist_ok=True)

# Convert every layer's dictionary to filename -> vector.
# If a basename occurs more than once within a layer, stop rather than
# silently choosing an arbitrary image.
by_layer = {}
for layer, d in edge_dicts.items():
    mapped = {}
    for original_key, vector in d.items():
        name = os.path.basename(original_key)
        if name in mapped:
            raise ValueError(f"Duplicate basename in {layer}: {name}")
        mapped[name] = vector
    by_layer[layer] = mapped

common = sorted(set.intersection(*(set(d) for d in by_layer.values())))
print("Matched edge images:", len(common))

if not common:
    raise ValueError("Still no common filenames. Share the diagnostic output.")

rows = []
for name in common:
    # e.g. airplane10_edges.png -> airplane
    stem = os.path.splitext(name)[0].lower()
    stem = re.sub(r"_edges$", "", stem)
    category = re.sub(r"\d+$", "", stem)

    rows.append({
        "image_id": name,
        "category": category,
        "y_knife": int(category == "knife")
    })

meta = pd.DataFrame(rows)
print("\nCategory counts:")
print(meta["category"].value_counts().to_string())
print("\nBinary label counts:")
print(meta["y_knife"].value_counts().rename(index={0:"non-knife", 1:"knife"}))

all_results = []

for layer, d in by_layer.items():
    X = np.asarray([d[name] for name in common], dtype=np.float32)
    y = meta["y_knife"].to_numpy()

    if np.bincount(y, minlength=2).min() < 5:
        print(f"Skipping {layer}: too few examples in one class.")
        continue

    p_sum = np.zeros(len(y))
    n_seen = np.zeros(len(y), dtype=int)
    cv = RepeatedStratifiedKFold(n_splits=5, n_repeats=5, random_state=42)

    for train_idx, test_idx in cv.split(X, y):
        model = LogisticRegression(
            class_weight="balanced", max_iter=5000, solver="lbfgs"
        )
        model.fit(X[train_idx], y[train_idx])
        knife_col = list(model.classes_).index(1)
        p_sum[test_idx] += model.predict_proba(X[test_idx])[:, knife_col]
        n_seen[test_idx] += 1

    p_knife = p_sum / n_seen
    pred = (p_knife >= 0.5).astype(int)

    result = meta.copy()
    result["condition"] = "Edges"
    result["representation"] = layer
    result["probe_p_knife_mean_oof"] = p_knife
    result["probe_p_nonknife_mean_oof"] = 1 - p_knife
    result["probe_prediction"] = np.where(pred == 1, "knife", "non-knife")
    result["probe_correct"] = (pred == y).astype(int)
    result["probe_confidence"] = np.maximum(p_knife, 1 - p_knife)
    result["probe_margin"] = np.abs(2 * p_knife - 1)
    result["n_oof_repeats"] = n_seen
    all_results.append(result)

combined = pd.concat(all_results, ignore_index=True)
save_path = f"{OUT}/RECONSTRUCTED_edges_per_image.csv"
combined.to_csv(save_path, index=False)

print("\nSaved:", save_path)
print("\nReconstructed accuracy:")
print(combined.groupby("representation")["probe_correct"].mean().round(4))
```

```python

# ================================================================
# FINAL ASSEMBLY: BMT FALSE-KNIFE PER-IMAGE DATASET
# Searches Drive, joins outcomes + probe evidence, saves final CSVs,
# and reports any missing matches without inventing data.
# ================================================================

import os
import glob
import pandas as pd
import numpy as np

ROOT = "/content/drive/MyDrive/false_knife_analysis"
OUT = os.path.join(ROOT, "FINAL_ASSEMBLED_DATASET")
os.makedirs(OUT, exist_ok=True)

# Mount Drive if this is a fresh Colab session
try:
    from google.colab import drive
    if not os.path.isdir("/content/drive/MyDrive"):
        drive.mount("/content/drive")
except Exception:
    pass

def find_file(filename, preferred=None):
    if preferred and os.path.exists(preferred):
        return preferred
    hits = glob.glob("/content/drive/MyDrive/**/" + filename, recursive=True)
    return hits[0] if hits else None

def read_csv_safe(path):
    if not path or not os.path.isfile(path):
        return None
    try:
        return pd.read_csv(path)
    except Exception as e:
        print(f"Could not read {path}: {e}")
        return None

def norm_id(x):
    """Normalize IDs for joining: remove folder, extension and whitespace."""
    x = os.path.basename(str(x).strip().lower())
    x = os.path.splitext(x)[0]
    for suffix in [
        "_silhouette", "_silhouettes", "_edges", "_outline",
        "_outlines", "_black_filled", "_noise_filled"
    ]:
        if x.endswith(suffix):
            x = x[:-len(suffix)]
    return x.strip()

def find_id_col(df):
    if df is None:
        return None
    for c in ["image_id", "imagename", "filename", "image", "image_name"]:
        if c in df.columns:
            return c
    return None

def add_join_id(df, id_col):
    df = df.copy()
    df["join_id"] = df[id_col].map(norm_id)
    return df

# ----------------------- Locate files ----------------------------

cue_out_path = find_file(
    "FINAL_cue_conflict_47.csv",
    os.path.join(ROOT, "per_image_final_audit", "FINAL_cue_conflict_47.csv")
)
edge_probe_path = find_file(
    "FINAL_edges_false_knife_probe.csv",
    os.path.join(ROOT, "per_image_final_audit", "FINAL_edges_false_knife_probe.csv")
)
sil_probe_path = find_file(
    "FINAL_silhouettes_false_knife_probe.csv",
    os.path.join(ROOT, "per_image_final_audit", "FINAL_silhouettes_false_knife_probe.csv")
)
cue_joined_path = find_file(
    "JOINED_cue_conflict_47_outcomes_plus_probe.csv",
    os.path.join(ROOT, "direct_linear_probe", "per_image_reconstruction",
                 "JOINED_cue_conflict_47_outcomes_plus_probe.csv")
)
trajectory_path = find_file(
    "false_knife_four_condition_trajectories.csv",
    os.path.join(ROOT, "false_knife_four_condition_trajectories.csv")
)

print("FILES LOCATED")
for name, path in [
    ("Cue-conflict audit", cue_out_path),
    ("Cue-conflict joined file", cue_joined_path),
    ("Edges probe", edge_probe_path),
    ("Silhouette probe", sil_probe_path),
    ("Four-condition trajectories", trajectory_path)
]:
    print(f"{name}: {path if path else 'NOT FOUND'}")

cue = read_csv_safe(cue_joined_path) or read_csv_safe(cue_out_path)
edge_probe = read_csv_safe(edge_probe_path)
sil_probe = read_csv_safe(sil_probe_path)
traj = read_csv_safe(trajectory_path)

# ------------------- Cue-conflict assembly -----------------------

cue_final = None
if cue is not None:
    cue_id = find_id_col(cue)
    if cue_id:
        cue_final = add_join_id(cue, cue_id)
        cue_final["condition_group"] = "cue_conflict"
        cue_final.to_csv(
            os.path.join(OUT, "FINAL_cue_conflict_47_outcomes_plus_probe.csv"),
            index=False
        )
        print(f"\nCue-conflict rows assembled: {len(cue_final)}")

# ---------------------- Edges assembly ---------------------------

edge_final = None
if edge_probe is not None:
    edge_id = find_id_col(edge_probe)
    if edge_id:
        edge_final = add_join_id(edge_probe, edge_id)
        edge_final["condition_group"] = "edges"
        edge_final.to_csv(
            os.path.join(OUT, "FINAL_edges_probe_evidence.csv"), index=False
        )
        print(f"Edges probe rows assembled: {len(edge_final)}")

# ------------------ Silhouette outcome matching -----------------
# The four-condition trajectory file is NOT assumed to be a
# silhouette-only dataset. It is joined by image ID, and unmatched
# silhouette cases remain explicitly marked as missing.

sil_final = None
sil_outcome_matches = 0
sil_unmatched_ids = []

if sil_probe is not None:
    sil_id = find_id_col(sil_probe)
    if sil_id:
        sil_final = add_join_id(sil_probe, sil_id)
        sil_final["condition_group"] = "silhouettes"

        if traj is not None and "imagename" in traj.columns:
            tr = add_join_id(traj, "imagename")

            # Keep only useful outcome columns from the trajectory file.
            wanted = [
                "join_id", "original", "outlines", "black_filled",
                "noise_filled", "trajectory", "ever_left_knife",
                "ever_correct", "ended_correct", "returned_to_knife",
                "actual_category", "category"
            ]
            keep = [c for c in wanted if c in tr.columns]
            tr = tr[keep].drop_duplicates("join_id", keep="first")

            # Avoid duplicate category fields after merging.
            sil_final = sil_final.merge(
                tr, on="join_id", how="left", suffixes=("", "_trajectory")
            )

            outcome_cols = [
                c for c in [
                    "original", "outlines", "black_filled", "noise_filled",
                    "trajectory", "ever_left_knife", "ever_correct",
                    "ended_correct", "returned_to_knife"
                ] if c in sil_final.columns
            ]
            if outcome_cols:
                sil_outcome_matches = int(
                    sil_final[outcome_cols].notna().any(axis=1).sum()
                )

            sil_unmatched_ids = sil_final.loc[
                sil_final[outcome_cols].isna().all(axis=1), sil_id
            ].astype(str).tolist() if outcome_cols else sil_final[sil_id].astype(str).tolist()

        else:
            sil_final["outcome_match_status"] = "trajectory_file_not_found"

        if "trajectory" in sil_final.columns:
            sil_final["outcome_match_status"] = np.where(
                sil_final["trajectory"].notna(),
                "matched_to_four_condition_trajectory",
                "no_matching_trajectory_found"
            )
        else:
            sil_final["outcome_match_status"] = "no_matching_trajectory_found"

        sil_final.to_csv(
            os.path.join(OUT, "FINAL_silhouettes_probe_plus_available_outcomes.csv"),
            index=False
        )
        print(f"\nSilhouette probe rows: {len(sil_final)}")
        print(f"Silhouette rows with matched trajectory: {sil_outcome_matches}")
        print(f"Silhouette IDs without a matched trajectory: {len(sil_unmatched_ids)}")
        if sil_unmatched_ids:
            print("Unmatched silhouette IDs:", ", ".join(sil_unmatched_ids))

# ------------------- Combined master table ----------------------
# Preserve separate conditions and their own columns. This is a
# convenience table, not a claim that all outcome fields are present.

combined = []
for df in [cue_final, edge_final, sil_final]:
    if df is not None:
        combined.append(df)

if combined:
    all_data = pd.concat(combined, ignore_index=True, sort=False)
    all_data.to_csv(
        os.path.join(OUT, "FINAL_ALL_CONDITIONS_ASSEMBLED.csv"), index=False
    )
    print(f"\nCombined table saved: {len(all_data)} rows")

# ------------------------ Audit ---------------------------------

audit_rows = []

def audit_add(dataset, df, expected_n, outcome_status, probe_status):
    if df is None:
        audit_rows.append({
            "dataset": dataset, "rows_found": 0, "expected_rows": expected_n,
            "unique_image_ids": 0, "outcome_status": "FILE MISSING",
            "probe_status": "FILE MISSING", "complete": False
        })
        return

    idc = find_id_col(df)
    unique_n = df[idc].map(norm_id).nunique() if idc else np.nan
    audit_rows.append({
        "dataset": dataset,
        "rows_found": len(df),
        "expected_rows": expected_n,
        "unique_image_ids": unique_n,
        "outcome_status": outcome_status,
        "probe_status": probe_status,
        "complete": bool(
            len(df) == expected_n and unique_n == expected_n
            and "missing" not in outcome_status.lower()
            and "missing" not in probe_status.lower()
        )
    })

audit_add(
    "Cue-conflict",
    cue_final,
    47,
    "Outcome fields available in source file" if cue_final is not None else "Missing",
    "Probe fields in source file" if cue_final is not None else "Missing"
)
audit_add(
    "Edges",
    edge_final,
    130,
    "Four-condition trajectories available separately",
    "Reconstructed probe evidence available" if edge_final is not None else "Missing"
)
audit_add(
    "Silhouettes",
    sil_final,
    51,
    f"{sil_outcome_matches}/51 matched to available four-condition trajectory"
    if sil_final is not None else "Missing",
    "Reconstructed probe evidence available" if sil_final is not None else "Missing"
)

audit = pd.DataFrame(audit_rows)
audit.to_csv(os.path.join(OUT, "FINAL_COMPLETENESS_AUDIT.csv"), index=False)

print("\n================ FINAL COMPLETENESS AUDIT ================")
print(audit.to_string(index=False))
print("\nSaved files in:", OUT)
print("\nIMPORTANT:")
print("- Reconstructed probe probabilities are not the original saved classifiers.")
print("- Four-condition trajectories are not automatically silhouette-only outcomes.")
print("- Missing matches remain missing; no outcomes have been fabricated.")
```

```python

# ================================================================
# FINAL BMT FALSE-KNIFE DATASET ASSEMBLY
# Cue-conflict outcomes + probe evidence
# Edges probe evidence
# Silhouette probe evidence + available four-condition trajectories
# Saves combined CSVs and a completeness audit.
# ================================================================

import os
import glob
import pandas as pd
import numpy as np

ROOT = "/content/drive/MyDrive/false_knife_analysis"
OUT = os.path.join(ROOT, "FINAL_ASSEMBLED_DATASET")
os.makedirs(OUT, exist_ok=True)

# Mount Google Drive if necessary
try:
    from google.colab import drive
    if not os.path.isdir("/content/drive/MyDrive"):
        drive.mount("/content/drive")
except Exception:
    pass

# -------------------------- Helpers -----------------------------

def find_file(filename, preferred_path=None):
    if preferred_path and os.path.isfile(preferred_path):
        return preferred_path

    matches = glob.glob(
        "/content/drive/MyDrive/**/" + filename,
        recursive=True
    )
    return matches[0] if matches else None


def read_csv_safe(path):
    if not path or not os.path.isfile(path):
        return None
    try:
        return pd.read_csv(path)
    except Exception as e:
        print(f"ERROR reading {path}: {e}")
        return None


def norm_id(value):
    """Normalize image IDs for matching."""
    value = os.path.basename(str(value).strip().lower())
    value = os.path.splitext(value)[0]

    suffixes = [
        "_silhouette", "_silhouettes", "_edges",
        "_outline", "_outlines", "_black_filled",
        "_noise_filled"
    ]
    changed = True
    while changed:
        changed = False
        for suffix in suffixes:
            if value.endswith(suffix):
                value = value[:-len(suffix)]
                changed = True

    return value.strip()


def find_id_col(df):
    if df is None:
        return None
    candidates = [
        "image_id", "imagename", "filename",
        "image_name", "image", "file"
    ]
    for col in candidates:
        if col in df.columns:
            return col
    return None


def add_join_id(df, id_col):
    result = df.copy()
    result["join_id"] = result[id_col].map(norm_id)
    return result


def report_file(label, path):
    print(f"{label}: {path if path else 'NOT FOUND'}")


# ------------------------ Locate files --------------------------

cue_joined_path = find_file(
    "JOINED_cue_conflict_47_outcomes_plus_probe.csv",
    os.path.join(
        ROOT, "direct_linear_probe", "per_image_reconstruction",
        "JOINED_cue_conflict_47_outcomes_plus_probe.csv"
    )
)

cue_audit_path = find_file(
    "FINAL_cue_conflict_47.csv",
    os.path.join(ROOT, "per_image_final_audit",
                 "FINAL_cue_conflict_47.csv")
)

edge_probe_path = find_file(
    "FINAL_edges_false_knife_probe.csv",
    os.path.join(ROOT, "per_image_final_audit",
                 "FINAL_edges_false_knife_probe.csv")
)

sil_probe_path = find_file(
    "FINAL_silhouettes_false_knife_probe.csv",
    os.path.join(ROOT, "per_image_final_audit",
                 "FINAL_silhouettes_false_knife_probe.csv")
)

trajectory_path = find_file(
    "false_knife_four_condition_trajectories.csv",
    os.path.join(ROOT, "false_knife_four_condition_trajectories.csv")
)

print("=" * 65)
print("INPUT FILES")
print("=" * 65)
report_file("Cue-conflict joined file", cue_joined_path)
report_file("Cue-conflict audit file", cue_audit_path)
report_file("Edges probe", edge_probe_path)
report_file("Silhouette probe", sil_probe_path)
report_file("Four-condition trajectories", trajectory_path)

# FIX: never use `or` directly with pandas DataFrames.
cue = read_csv_safe(cue_joined_path)
if cue is None:
    cue = read_csv_safe(cue_audit_path)

edge_probe = read_csv_safe(edge_probe_path)
sil_probe = read_csv_safe(sil_probe_path)
trajectories = read_csv_safe(trajectory_path)


# -------------------- Cue-conflict assembly ---------------------

cue_final = None

if cue is not None:
    cue_id_col = find_id_col(cue)

    if cue_id_col:
        cue_final = add_join_id(cue, cue_id_col)
        cue_final["condition_group"] = "cue_conflict"

        cue_outfile = os.path.join(
            OUT, "FINAL_cue_conflict_outcomes_plus_probe.csv"
        )
        cue_final.to_csv(cue_outfile, index=False)

        print("\nCue-conflict table:")
        print("Rows:", len(cue_final))
        print("Unique image IDs:", cue_final["join_id"].nunique())
        print("Saved:", cue_outfile)
    else:
        print("\nWARNING: no recognizable image ID column in cue-conflict file.")
else:
    print("\nWARNING: cue-conflict input file could not be loaded.")


# ------------------------- Edges assembly -----------------------

edge_final = None

if edge_probe is not None:
    edge_id_col = find_id_col(edge_probe)

    if edge_id_col:
        edge_final = add_join_id(edge_probe, edge_id_col)
        edge_final["condition_group"] = "edges"

        edge_outfile = os.path.join(
            OUT, "FINAL_edges_probe_evidence.csv"
        )
        edge_final.to_csv(edge_outfile, index=False)

        print("\nEdges probe table:")
        print("Rows:", len(edge_final))
        print("Unique image IDs:", edge_final["join_id"].nunique())
        print("Saved:", edge_outfile)
    else:
        print("\nWARNING: no recognizable image ID column in edge probe file.")
else:
    print("\nWARNING: edge probe file could not be loaded.")


# -------------------- Silhouette assembly -----------------------
# We attach the available four-condition trajectories by normalized ID.
# The source trajectory file is NOT assumed to be silhouette-specific.
# Unmatched cases are preserved and marked.

sil_final = None
silhouette_match_count = 0
silhouette_unmatched = []

if sil_probe is not None:
    sil_id_col = find_id_col(sil_probe)

    if sil_id_col:
        sil_final = add_join_id(sil_probe, sil_id_col)

        # Identify the probe columns that should be present.
        sil_final["condition_group"] = "silhouettes"

        if trajectories is not None and "imagename" in trajectories.columns:
            tr = add_join_id(trajectories, "imagename")

            trajectory_cols = [
                "join_id",
                "original",
                "outlines",
                "black_filled",
                "noise_filled",
                "trajectory",
                "ever_left_knife",
                "ever_correct",
                "ended_correct",
                "returned_to_knife",
                "actual_category",
                "category"
            ]
            keep_cols = [
                col for col in trajectory_cols if col in tr.columns
            ]

            # One record per normalized ID to avoid multiplying rows.
            tr = tr[keep_cols].drop_duplicates(
                subset="join_id", keep="first"
            )

            # Keep probe rows even when no outcome match exists.
            sil_final = sil_final.merge(
                tr,
                on="join_id",
                how="left",
                suffixes=("", "_trajectory"),
                validate="many_to_one"
            )

            outcome_cols = [
                col for col in [
                    "original", "outlines", "black_filled",
                    "noise_filled", "trajectory", "ever_correct",
                    "ended_correct"
                ]
                if col in sil_final.columns
            ]

            if outcome_cols:
                matched_mask = sil_final[outcome_cols].notna().any(axis=1)
                silhouette_match_count = int(matched_mask.sum())
                silhouette_unmatched = (
                    sil_final.loc[~matched_mask, sil_id_col]
                    .astype(str).tolist()
                )
            else:
                matched_mask = pd.Series(
                    False, index=sil_final.index
                )

            sil_final["outcome_match_status"] = np.where(
                matched_mask,
                "matched_to_four_condition_trajectory",
                "no_matching_trajectory_found"
            )

        else:
            sil_final["outcome_match_status"] = (
                "trajectory_file_missing_or_no_image_ID"
            )
            silhouette_unmatched = sil_final[sil_id_col].astype(str).tolist()

        sil_outfile = os.path.join(
            OUT, "FINAL_silhouettes_probe_plus_available_outcomes.csv"
        )
        sil_final.to_csv(sil_outfile, index=False)

        print("\nSilhouette table:")
        print("Probe rows:", len(sil_final))
        print("Unique image IDs:", sil_final["join_id"].nunique())
        print("Rows with a trajectory match:", silhouette_match_count)
        print("Rows without a trajectory match:", len(silhouette_unmatched))
        if silhouette_unmatched:
            print("Unmatched IDs:", ", ".join(silhouette_unmatched))
        print("Saved:", sil_outfile)

    else:
        print("\nWARNING: no recognizable image ID column in silhouette probe file.")
else:
    print("\nWARNING: silhouette probe file could not be loaded.")


# ----------------------- Combined dataset -----------------------
# Each condition retains its own available columns.
# This does not imply every condition has identical outcome measures.

available_tables = [
    df for df in [cue_final, edge_final, sil_final]
    if df is not None
]

combined = None

if available_tables:
    combined = pd.concat(
        available_tables,
        ignore_index=True,
        sort=False
    )

    combined_outfile = os.path.join(
        OUT, "FINAL_ALL_CONDITIONS_ASSEMBLED.csv"
    )
    combined.to_csv(combined_outfile, index=False)

    print("\nCombined dataset:")
    print("Rows:", len(combined))
    if "condition_group" in combined.columns:
        print(combined["condition_group"].value_counts(dropna=False).to_string())
    print("Saved:", combined_outfile)


# ---------------------- Completeness audit ----------------------

audit_rows = []

def add_audit(dataset, df, expected_n, outcome_status, probe_status):
    if df is None:
        audit_rows.append({
            "dataset": dataset,
            "rows_found": 0,
            "expected_rows": expected_n,
            "unique_image_ids": 0,
            "outcome_status": "FILE MISSING",
            "probe_status": "FILE MISSING",
            "complete": False
        })
        return

    id_col = find_id_col(df)
    unique_count = (
        df[id_col].map(norm_id).nunique()
        if id_col else np.nan
    )

    audit_rows.append({
        "dataset": dataset,
        "rows_found": len(df),
        "expected_rows": expected_n,
        "unique_image_ids": unique_count,
        "outcome_status": outcome_status,
        "probe_status": probe_status,
        "complete": bool(
            len(df) == expected_n
            and unique_count == expected_n
            and "missing" not in str(outcome_status).lower()
            and "unmatched" not in str(outcome_status).lower()
            and "missing" not in str(probe_status).lower()
        )
    })


add_audit(
    "Cue-conflict",
    cue_final,
    47,
    "Outcome/probe source loaded" if cue_final is not None else "Missing",
    "Probe data included" if cue_final is not None else "Missing"
)

add_audit(
    "Edges",
    edge_final,
    130,
    "Four-condition trajectory is a separate source, not merged here",
    "Probe data loaded" if edge_final is not None else "Missing"
)

sil_outcome_status = (
    f"{silhouette_match_count}/{len(sil_final)} probe rows matched to "
    "four-condition trajectory"
    if sil_final is not None else "Missing"
)

add_audit(
    "Silhouettes",
    sil_final,
    51,
    sil_outcome_status,
    "Probe data loaded" if sil_final is not None else "Missing"
)

audit = pd.DataFrame(audit_rows)
audit_path = os.path.join(OUT, "FINAL_COMPLETENESS_AUDIT.csv")
audit.to_csv(audit_path, index=False)

print("\n" + "=" * 65)
print("FINAL COMPLETENESS AUDIT")
print("=" * 65)
print(audit.to_string(index=False))

print("\nAll output files saved in:")
print(OUT)

print("\nINTERPRETATION NOTES")
print("1. Probe probabilities are from the reconstructed probe analysis.")
print("2. Four-condition trajectories are not automatically silhouette-only outcomes.")
print("3. Unmatched images remain unmatched; no outcomes are fabricated.")
print("4. Edges and silhouettes have separate probe evidence files.")
```

```python

import os
import json
import pandas as pd

ROOTS = [
    "/content/drive/MyDrive/false_knife_analysis",
    "/content/drive/MyDrive/AI_Mistakes_Error_Framework"
]

KEYWORDS = [
    "silhouette", "unmask", "progressive", "restoration",
    "occlusion", "patch", "trajectory", "prediction",
    "result", "summary", "stage", "intervention",
    "step_3", "step_2", "trial_data"
]

print("=" * 75)
print("FINAL SEARCH: SILHOUETTE UNMASKING + EDGE PATCH OCCLUSION")
print("=" * 75)

candidates = []

for root in ROOTS:
    if not os.path.isdir(root):
        print("\nFolder missing:", root)
        continue

    for folder, dirs, files in os.walk(root):
        dirs[:] = [
            d for d in dirs
            if d not in {".git", "node_modules", "__pycache__"}
        ]

        for filename in files:
            ext = os.path.splitext(filename)[1].lower()
            if ext not in {".csv", ".json", ".xlsx", ".pkl", ".parquet"}:
                continue

            path = os.path.join(folder, filename)
            lower = filename.lower()

            if any(word in lower for word in KEYWORDS):
                candidates.append(path)

print(f"\nCandidate files found: {len(candidates)}")

for path in sorted(set(candidates)):
    print(path)

print("\n" + "=" * 75)
print("CHECKING CSV/JSON CONTENTS FOR MODEL PREDICTION EVIDENCE")
print("=" * 75)

prediction_terms = [
    "predicted_category", "predicted_class", "prediction",
    "pred_label", "top1", "top_1", "alpha", "stage",
    "patch_idx", "patch_index", "patch_prediction",
    "correct_category", "is_correct", "imagename", "image_id"
]

for path in sorted(set(candidates)):
    ext = os.path.splitext(path)[1].lower()

    try:
        if ext == ".csv":
            # Read headers first, avoiding a full load of large CSVs.
            headers = pd.read_csv(path, nrows=0).columns.tolist()
            relevant = [
                col for col in headers
                if any(term in col.lower() for term in prediction_terms)
            ]

            # Only inspect files with potentially useful prediction columns.
            if relevant:
                df = pd.read_csv(path, nrows=3)
                print("\nFILE:", path)
                print("Columns:", headers)
                print("Relevant columns:", relevant)
                print(df.to_string(index=False, max_cols=15))

        elif ext == ".json" and os.path.getsize(path) < 15_000_000:
            with open(path, "r", encoding="utf-8") as f:
                obj = json.load(f)

            if isinstance(obj, dict):
                keys = list(obj.keys())
                print("\nJSON:", path)
                print("Top-level keys (first 12):", keys[:12])

                # Show a sample value structure, not the entire dataset.
                if keys:
                    sample = obj[keys[0]]
                    if isinstance(sample, dict):
                        print("Sample value keys:", list(sample.keys())[:20])
                    elif isinstance(sample, list):
                        print("Sample value:", sample[:3])
                    else:
                        print("Sample value:", str(sample)[:180])

    except Exception as e:
        print("\nCould not inspect:", path)
        print("Reason:", str(e)[:200])

print("\n" + "=" * 75)
print("END OF FINAL SEARCH")
print("No experiments rerun. No files modified.")
print("=" * 75)
```
